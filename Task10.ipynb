#### 5. Find the factorial of a number using for loop and also using while loop.

### Using for loop 

```python
num=int(input("Enter a number :"))
factorial=1

if num<0:
    print("Negative number does not exist")

elif num==0:
    print("Factorial of 0 is 1")
else:
    for i in range(1,num+1):
        factorial*=i
    print("Factorial :",factorial)
```

### Using While loop

```python
num=int(input("Enter a number"))
factorial=1
i=1

while i <= num:
    factorial*=i
    i=i+1

print("Factorial :",factorial)
```

#### 10. Create a class Student with private attribute __marks. Add getter and setter with validation (0–100).

```python
class Student:
    def __init__(self, marks):
        self.__marks = 0
        self.set_marks(marks)

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks! Marks should be between 0 and 100.")

    def get_marks(self):
        return self.__marks

student = Student(76)

print("Marks:", student.get_marks())

student.set_marks(85)
print("Updated Marks:", student.get_marks())

student.set_marks(120)
print("Final Marks:", student.get_marks())
```

#### 11. Create class BankAccount with private __balance. Add deposit and withdraw methods. Raise custom exception if balance becomes negative.

```python
class InsufficientBalanceError(Exception):
    pass

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Deposited:", amount)
        print("Current Balance:", self.__balance)

    def withdraw(self, amount):
        if self.__balance - amount < 0:
            raise InsufficientBalanceError("Insufficient Balance!")
        else:
            self.__balance -= amount
            print("Withdrawn:", amount)
            print("Current Balance:", self.__balance)

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)

try:
    account.deposit(3000)
    account.withdraw(4000)
    account.withdraw(7000)
except InsufficientBalanceError as e:
    print("Error:", e)

print("Final Balance:", account.get_balance())
```

#### 13. Create class Product with private __price. Add getter and setter. Setter should not allow price less than 100.

```python
class Product:
    def __init__(self, price):
        self.__price = 100
        self.set_price(price)

    def set_price(self, price):
        if price >= 100:
            self.__price = price
        else:
            print("Invalid Price! Price should be at least 100.")

    def get_price(self):
        return self.__price

product = Product(700)

print("Price:", product.get_price())

product.set_price(900)
print("Updated Price:", product.get_price())

product.set_price(50)

print("Final Price:", product.get_price())
```

#### 14. Create class Rectangle with private length and width. Add method to calculate area and perimeter.

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length
        self.__width = width

    def area(self):
        return self.__length * self.__width

    def perimeter(self):
        return 2 * (self.__length + self.__width)

rect = Rectangle(10, 5)

print("Area:", rect.area())
print("Perimeter:", rect.perimeter())
```

# Inheritance + Abstraction + super()

#### 15. Create single level inheritance: Person → Student. Use super() in constructor.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Student(Person):
    def __init__(self, name, age, course):
        super().__init__(name, age)
        self.course = course

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)

student1 = Student("Yash", 22, "B.Tech CSE")

student1.display()
```

#### 16. Create multi-level inheritance: Person → Employee → Manager. Override a method at each level.

```python
class Person:
    def show(self):
        print("I am a Person.")

class Employee(Person):
    def show(self):
        print("I am an Employee.")

class Manager(Employee):
    def show(self):
        print("I am a Manager.")

p = Person()
e = Employee()
m = Manager()

p.show()
e.show()
m.show()
```

#### 17. Create hierarchical inheritance: Parent Animal → Dog and Cat. Override sound() method.

```python
class Animal:
    def sound(self):
        print("Animal makes a sound.")

class Dog(Animal):
    def sound(self):
        print("Dog says: Bark Bark")

class Cat(Animal):
    def sound(self):
        print("Cat says: Meow Meow")

dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

#### 18. Create abstract class Shape with abstract method area(). Implement it in Circle and Square using super().

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self, value):
        self.value = value

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        super().__init__(radius)

    def area(self):
        return 3.14 * self.value * self.value

class Square(Shape):
    def __init__(self, side):
        super().__init__(side)

    def area(self):
        return self.value * self.value

circle = Circle(8)
square = Square(5)

print("Area of Circle:", circle.area())
print("Area of Square:", square.area())
```

#### 20. Create a class Calculator that can add 2 or 3 numbers (method overloading using *args).

```python
class Calculator:
    def add(self, *args):
        if len(args) == 2:
            print("Sum of 2 numbers:", args[0] + args[1])
        elif len(args) == 3:
            print("Sum of 3 numbers:", args[0] + args[1] + args[2])
        else:
            print("Please enter only 2 or 3 numbers.")

calc = Calculator()

calc.add(10, 20)
calc.add(10, 20, 30)
calc.add(10)
```
