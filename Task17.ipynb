# Matplotlib

####  1. Create a simple line plot using:
####  o X = [1, 2, 3, 4, 5]
####  o Y = [10, 20, 15, 25, 30]

```python
import matplotlib.pyplot as plt

X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)

plt.show()
```

#### 2. Create a line plot and add an appropriate title.

```python
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.title("Simple Line Plot")
plt.show()
```

#### 3. Create a line plot and label both the X-axis and Y-axis.

```python
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

#### 4. Plot two lines on the same graph and add a legend to identify them.

```python
X = [1, 2, 3, 4, 5]

Y1 = [10, 20, 15, 25, 30]
Y2 = [5, 15, 20, 22, 35]

plt.plot(X, Y1, label="Line 1")
plt.plot(X, Y2, label="Line 2")
plt.title("Comparison of Two Lines")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

#### 5. Change the line color, line style, and marker style of a line plot.

```python
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(
    X, Y,
    color="Blue",
    linestyle="-",
    marker="o"
)

plt.title("Customized Line Plot")
plt.xlabel("X")
plt.ylabel("Y")

plt.show()
```

#### 6. Create a line plot using NumPy values generated from np.arange().

```python
import numpy as np
X = np.arange(1, 6)
Y = np.arange(10, 60, 10)

plt.plot(X, Y)
plt.title("Line Plot ")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

#### 7. Create a bar chart showing the sales of 5 different products.

#### 10.Create a histogram showing the distribution of 100 randomly generated values.

```python
import numpy as np
import matplotlib.pyplot as plt

data=np.random.randn(100)
plt.hist(data, bins=10)
plt.title("Distribution of 100 random values")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```

#### 11.Create a pie chart showing the percentage of customers from 4 different cities.

```python
cities = ["Bangalore", "Nagpur", "Pune", "Hyderabad"]
customers = [40, 35, 30, 45]

plt.pie(customers, labels=cities,autopct="%1.1f%%")

plt.title("Percentage of Customers in Different cities")
plt.show()
```

#### 12.Create a plot containing two subplots:
#### o First subplot → line plot
#### o Second subplot → scatter plot

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(x, y, marker='o')
axes[0].set_title("Line Plot")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")
axes[0].grid(True)

axes[1].scatter(x, y)
axes[1].set_title("Scatter Plot")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Y")
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

#### 13.Create a graph for the mathematical function: for values of x from -10 to 10.

```python
x = np.arange(-10, 11)
y = x**2

plt.plot(x, y, marker='o')
plt.title("Graph of y = x²")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)

plt.show()
```

#### 14.Create a graph containing three mathematical functions: Use different markers/line styles and add a legend.

```python
x = np.arange(-10, 11)

y1 = x**2
y2 = x**3
y3 = 2*x + 1

plt.plot(x, y1, 'o-', label='y = x²')
plt.plot(x, y2, 's--', label='y = x³')
plt.plot(x, y3, '^-.' , label='y = 2x + 1')
plt.title("Three Functions")
plt.xlabel("x")
plt.ylabel("y")

plt.legend()
plt.grid(True)
plt.show()
```

#### 15.Create a complete business-style visualization using Matplotlib:
#### o X-axis → Month
#### o Y-axis → Sales
#### o Plot monthly sales
#### o Add title
#### o Add X/Y labels
#### o Add grid
#### o Add markers
#### o Add legend
#### o Customize figure size

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

sales = [25000, 28000, 32000, 30000, 35000, 38000,
         42000, 40000, 45000, 48000, 52000, 55000]

plt.figure(figsize=(12, 6))

plt.plot(
    months,
    sales,
    marker='o',
    linestyle='-',
    label='Monthly Sales'
)

plt.title("Monthly Sales Report", fontsize=16)
plt.xlabel("Month", fontsize=12)
plt.ylabel("Sales (₹)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()
plt.show()
```
