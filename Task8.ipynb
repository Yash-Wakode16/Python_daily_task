# File Handling



#### Basic File Handling

#### 1. Write a program to create a text file and write your name and age into it.

```python
file=open("person.txt","w")

file.write("Name : Yash Wakode\n")
file.write("Age : 22")

file.close()
print("Written successfully")
```

#### 2. Read the content of a text file and print it on the screen.

```python
file=open("person.txt","r")

print(file.read())

file.close()
```

#### 3. Write a program to append some text to an existing file.

```python
file=open("person.txt","a")
file.write("\nCity: Hyderabad")
file.write("\nCourse : Data Science")

file.close()
print("Data appended")
```

#### 4. Count the number of lines in a text file.

```python
file = open("person.txt", "r")
count = 0
for line in file:
    count += 1

print("Number of lines:", count)

file.close()
```

#### 5. Count the number of words in a text file.

```python
file=open("person.txt","r")
text=file.read()
words=text.split()
count=0
for i in words:
    count+=1
print("Total Words",count)
file.close()
```

#### 6. Count the number of characters in a text file.

```python
file=open("person.txt","r")
text=file.read()
count=0
for i in text:
    count+=1

print("Total characters :",count)
file.close()
```

#### 7. Read a file line by line using a loop.

```python
file = open("person.txt", "r")

for line in file:
    print(line, end="")

file.close()
```

#### 8. Write a program to copy content from one file to another file.

```python
source = open("person.txt", "r")
destination = open("destination.txt", "w")
content = source.read()

destination.write(content)
source.close()
destination.close()

print("Content copied successfully!")
```

#### 9. Check if a file exists or not before reading it.

```python
import os
filename = "sample.txt"
if os.path.exists(filename):
    file = open(filename, "r")
    print(file.read())
    file.close()
else:
    print("File does not exist.")
```

#### 10. Write a program to delete a file (use os module).

```python
import os
filename = "sample.txt"
if os.path.exists(filename):
    os.remove(filename)
    print("File deleted successfully.")
else:
    print("File does not exist.")
```

#### 11. Take student name and marks from user and store them in a file.

```python
file = open("student.txt", "w")
name = input("Enter student name: ")
marks = input("Enter student marks: ")

file.write("Student Name: " + name + "\n")
file.write("Marks: " + marks + "\n")

file.close()

print("Student data saved successfully!")
```
