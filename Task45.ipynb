#### 1. Print numbers from 1 to 100 using a for loop.

```python
for i in range(1,101):
    print(i)
```

#### 2. Print all even numbers from 1 to 100.

```python
for i in range(2,101,2):
    print(i)
```

#### 3. Find the sum of all even numbers from 1 to N.

```python
n=int(input())
total=0
for i in range(2,n+1,2):
    total=total+i
print(total)
```

#### 4. Find the sum of digits of a number.

```python
n=int(input())
total=0
while n>0:
    digit=n%10
    total=total+digit
    n=n//10
print(total)
```
