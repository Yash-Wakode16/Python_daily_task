#### 1. Class Name: BankAccount
#### Problem
#### Create a class named BankAccount.
#### Explanation
#### A bank account stores money. The balance should not be directly accessible from outside
#### the class, so create a private attribute named __balance.
#### Implement the following methods:
#### • deposit(amount)
#### o Deposit money into the account.
#### o Amount must be greater than 0.
#### • withdraw(amount)
#### o Withdraw money.
#### o Withdrawal should happen only if:
#### ▪ amount > 0
#### ▪ amount <= available balance

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"₹{amount} deposited successfully.")
        else:
            print("Deposit amount must be greater than 0.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be greater than 0.")
        elif amount <= self.__balance:
            self.__balance -= amount
            print(f"₹{amount} withdrawn successfully.")
        else:
            print("Insufficient balance.")

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)
account.deposit(2000)
print("Balance:", account.get_balance())
account.withdraw(1000)
print("Balance:", account.get_balance())
account.withdraw(10000)
```

#### 2. Class Name: Student
#### Problem
#### Create a class named Student.
#### Explanation
#### Every student has marks that should be private.
#### Create:
#### • Private attribute __marks
#### Methods: 
#### • set_marks(marks)
#### o Accept marks only between 0 and 100.
#### • get_marks()
#### o Return the student's marks.

```python
class Student:
    def __init__(self, marks=0):
        self.__marks = marks

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
            print("Marks updated successfully.")
        else:
            print("Marks must be between 0 and 100.")

    def get_marks(self):
        return self.__marks
student = Student()
student.set_marks(85)
print("Student Marks:", student.get_marks())
student.set_marks(120)
print("Student Marks:", student.get_marks())
```

#### 3. Class Name: Employee
#### Problem
#### Create a class named Employee.
#### Explanation
#### Salary should not be modified directly.
#### Create:
#### • Private attribute __salary
#### Methods:
#### • set_salary(new_salary)
#### o Salary cannot decrease by more than 10% of the current salary.
#### • get_salary()
#### Example:
#### Current Salary = 50,000
#### Allowed minimum = 45,000
#### New salary = 40,000
#### Reject

```python
class Employee:
    def __init__(self, salary):
        self.__salary = salary

    def set_salary(self, new_salary):
        minimum_salary = self.__salary * 0.90

        if new_salary >= minimum_salary:
            self.__salary = new_salary
            print("Salary updated successfully.")
        else:
            print("Salary cannot decrease by more than 10%.")

    def get_salary(self):
        return self.__salary
employee = Employee(50000)
print("Current Salary:", employee.get_salary())
employee.set_salary(46000)
print("New Salary:", employee.get_salary())
employee.set_salary(40000)
print("Salary:", employee.get_salary())
```

#### 6. Class Name: Product
#### Problem
#### Create a class named Product.
#### Explanation
#### Store
#### • private __price
#### • private __stock
#### Methods:
#### • update_price(new_price)
#### • update_stock(quantity)
#### Validation:
#### Price cannot be negative.
#### Stock cannot become negative.

```python
class Product:
    def __init__(self, price=0, stock=0):
        self.__price = price
        self.__stock = stock

    def update_price(self, new_price):
        if new_price >= 0:
            self.__price = new_price
            print("Price updated successfully.")
        else:
            print("Price cannot be negative.")

    def update_stock(self, quantity):
        if self.__stock + quantity >= 0:
            self.__stock += quantity
            print("Stock updated successfully.")
        else:
            print("Stock cannot become negative.")

    def get_price(self):
        return self.__price

    def get_stock(self):
        return self.__stock
product = Product(1000, 10)
print("Price:", product.get_price())
print("Stock:", product.get_stock())
product.update_price(1200)
product.update_stock(5)
print("Updated Price:", product.get_price())
print("Updated Stock:", product.get_stock())
product.update_price(-500)
product.update_stock(-20)
```

#### 7. Class Name: Car
#### Problem
#### Create class Car.
#### Explanation
#### Store privately
#### • __model
#### • __year
#### Provide
#### • get_model()
#### • set_model()
#### • get_year()
#### • set_year()
#### Validation:
#### Year cannot be greater than current year.

```python
from datetime import datetime

class Car:
    def __init__(self, model, year):
        self.__model = model
        self.__year = year

    def get_model(self):
        return self.__model

    def set_model(self, model):
        self.__model = model

    def get_year(self):
        return self.__year

    def set_year(self, year):
        current_year = datetime.now().year

        if year <= current_year:
            self.__year = year
            print("Year updated successfully.")
        else:
            print("Year cannot be greater than the current year.")

car = Car("Toyota", 2020)
print("Model:", car.get_model())
print("Year:", car.get_year())
car.set_model("Honda")
car.set_year(2022)
print("Updated Model:", car.get_model())
print("Updated Year:", car.get_year())
car.set_year(2030)
```

#### 8. Class Name: User
#### Problem
#### Create class User.
#### Explanation
#### Password must remain private.
#### Private attribute
#### __password
#### Method
#### change_password(new_password)
#### Validation
#### Password length ≥ 8

```python
class User:
    def __init__(self, password):
        self.__password = password

    def change_password(self, new_password):
        if len(new_password) >= 8:
            self.__password = new_password
            print("Password changed successfully.")
        else:
            print("Password must contain at least 8 characters.")

user = User("password123")
user.change_password("python123")
user.change_password("hello")
```

#### 9. Class Name: Rectangle
#### Problem
#### Create class Rectangle.
#### Explanation
#### Private attributes
#### • __length
#### • __width
#### Methods
#### • set_length()
#### • get_length()
#### • set_width()
#### • get_width()
#### • area()
#### Validation
#### Length and width must be positive.

```python
class Rectangle:
    def __init__(self, length=1, width=1):
        self.__length = length
        self.__width = width

    def set_length(self, length):
        if length > 0:
            self.__length = length
            print("Length updated successfully.")
        else:
            print("Length must be positive.")

    def get_length(self):
        return self.__length

    def set_width(self, width):
        if width > 0:
            self.__width = width
            print("Width updated successfully.")
        else:
            print("Width must be positive.")

    def get_width(self):
        return self.__width

    def area(self):
        return self.__length * self.__width
rectangle = Rectangle()
rectangle.set_length(10)
rectangle.set_width(5)
print("Length:", rectangle.get_length())
print("Width:", rectangle.get_width())
print("Area:", rectangle.area())
rectangle.set_length(-10)
rectangle.set_width(0)
```

#### 10. Class Name: Account
#### Problem
#### Create class Account.
#### Explanation
#### Private attribute
#### __balance
#### Whenever balance changes,
#### store every balance value inside a class variable
#### Example
#### history = []
#### Methods
#### • deposit()
#### • withdraw()
#### Every successful update should append the latest balance to history.

```python
class Account:
    history = []

    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            Account.history.append(self.__balance)
            print("Deposit successful.")
        else:
            print("Amount must be greater than 0.")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            Account.history.append(self.__balance)
            print("Withdrawal successful.")
        else:
            print("Invalid withdrawal.")

    def get_balance(self):
        return self.__balance
account = Account(5000)
print("Initial Balance:", account.get_balance())
account.deposit(2000)
print("Balance:", account.get_balance())
account.withdraw(1000)
print("Balance:", account.get_balance())
account.withdraw(10000)

print("History:", Account.history)
```
