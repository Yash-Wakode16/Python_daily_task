#### 1. Take a list of numbers and find the second largest number without using sort().

```python
numbers = [10, 25, 5, 40, 30]
largest = second_largest = float('-inf')
for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print("Second largest number:", second_largest)
```

```python
#### 3. Print all prime numbers between 1 and 100.
```

```python
for num in range(2, 101):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

```python
#### 5. Remove duplicate elements from a list without using set().
```

```python
numbers = [10, 20, 10, 30, 20, 40, 30]
unique = []
for num in numbers:
    if num not in unique:
        unique.append(num)
print(unique)
```

#### 8. Take a sentence and create a dictionary containing the frequency of each word. Example: "python is easy python is powerful"

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1
print(frequency)
```
