# Multiple Inheritance

#### 1. Create two parent classes Father and Mother. Create a child class Child that inherits from both.

```python
class Father:
    def father_property(self):
        print("Father's property")
class Mother:
    def mother_property(self):
        print("Mother's property")
class Child(Father, Mother):
    def child_info(self):
        print("Child")
c = Child()
c.father_property()
c.mother_property()
c.child_info()
```

#### 2. Define class A with method show() and class B with method display(). Create class C that inherits from both and calls both methods.

```python
class A:
    def show(self):
        print("This is class A")
class B:
    def display(self):
        print("This is class B")
class C(A, B):
    pass
obj = C()
obj.show()
obj.display()
```

#### 3. Create classes Teacher and Author. Create class Person that inherits from both and has attributes from both.

```python
class Teacher:
    def __init__(self, subject):
        self.subject = subject
class Author:
    def __init__(self, book):
        self.book = book
class Person(Teacher, Author):
    def __init__(self, subject, book):
        Teacher.__init__(self, subject)
        Author.__init__(self, book)
p = Person("Python", "Learn Python")
print("Subject:", p.subject)
print("Book:", p.book)
```

#### 4. Define class Sports and class Academic. Create class Student that inherits from both.

```python
class Sports:
    def sports_activity(self):
        print("Student plays cricket")
class Academic:
    def academic_activity(self):
        print("Student studies Python")
class Student(Sports, Academic):
    pass
s = Student()
s.sports_activity()
s.academic_activity()
```

#### 5. Create class Camera and class MusicPlayer. Create class Smartphone that inherits from both.

```python
class Camera:
    def take_photo(self):
        print("Taking a photo")
class MusicPlayer:
    def play_music(self):
        print("Playing music")
class Smartphone(Camera, MusicPlayer):
    pass
phone = Smartphone()
phone.take_photo()
phone.play_music()
```

#### 6. Define two classes with same method name. Create a child class that inherits from both and resolve the method conflict.

```python
class Father:
    def show(self):
        print("Father's method")
class Mother:
    def show(self):
        print("Mother's method")
class Child(Father, Mother):
    def show(self):
        print("Child's method")
c = Child()
c.show()
```

#### 7. Create class Engine and class Wheels. Create class Car that inherits from both.

```python
class Engine:
    def start_engine(self):
        print("Engine started")
class Wheels:
    def rotate_wheels(self):
        print("Wheels are rotating")
class Car(Engine, Wheels):
    pass
car = Car()
car.start_engine()
car.rotate_wheels()
```

#### 8. Define class Writer and class Singer. Create class Artist that inherits from both.

```python
class Writer:
    def write(self):
        print("Artist can write")
class Singer:
    def sing(self):
        print("Artist can sing")
class Artist(Writer, Singer):
    pass
a = Artist()
a.write()
a.sing()
```

#### 9. Create class Bank and class Insurance. Create class Customer that inherits from both.

```python
class Bank:
    def bank_service(self):
        print("Bank service available")


class Insurance:
    def insurance_service(self):
        print("Insurance service available")


class Customer(Bank, Insurance):
    pass


# Create object
c = Customer()

c.bank_service()
c.insurance_service()
```
