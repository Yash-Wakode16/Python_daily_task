#### 1. Take a number and find the sum of its digits. Example: 1234 → 10

```python
num = int(input("Enter a number: "))
sum_digits = 0
while num > 0:
    digit = num % 10
    sum_digits += digit
    num = num // 10
print("Sum of digits:", sum_digits)
```

#### 2. Take a number and count how many digits it contains.

```python
num = int(input("Enter a number: "))
count = 0
while num > 0:
    num = num // 10
    count += 1
print("Number of digits:", count)
```

#### 3. Reverse a number without converting it to a string. Example: 12345 → 54321

```python
num = int(input("Enter a number: "))
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
print("Reversed number:", reverse)
```

#### 4. Check whether a number is an Armstrong number. Example: 153 → Armstrong


```python
num = int(input("Enter a number: "))
original = num
digits = 0
temp = num
while temp > 0:
    temp = temp // 10
    digits += 1

sum_digits = 0
temp = num
while temp > 0:
    digit = temp % 10
    sum_digits += digit ** digits
    temp = temp // 10
if sum_digits == original:
    print("Armstrong")
else:
    print("Not Armstrong")
```

#### 5. Print the first n terms of the Fibonacci series. Example: 0 1 1 2 3 5 8

```python
n = int(input("Enter the number of terms: "))
a = 0
b = 1
for i in range(n):
    print(a, end=" ")
    a, b = b, a + b
```

#### 6. Take a number and print its multiplication table from 1 to 10.

```python
num = int(input("Enter a number: "))
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

#### 7. Check whether a number is a perfect number. Example: 6 → 1 + 2 + 3 = 6

```python
num = int(input("Enter a number: "))
sum_divisors = 0
for i in range(1, num):
    if num % i == 0:
        sum_divisors += i
if sum_divisors == num:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

#### 8. Take a number and print all of its factors.

```python
num = int(input("Enter a number: "))
print("Factors are:")
for i in range(1, num + 1):
    if num % i == 0:
        print(i, end=" ")
```

#### 9. Take two numbers and print all prime numbers between them.

```python
start = int(input("Enter the first number: "))
end = int(input("Enter the second number: "))
print("Prime numbers:")
for num in range(start, end + 1):
    if num < 2:
        continue
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num, end=" ")
```

#### 10. Find the Greatest Common Divisor (GCD) of two numbers without using math.gcd().

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
while num2 != 0:
    num1, num2 = num2, num1 % num2
print("GCD:", num1)
```
