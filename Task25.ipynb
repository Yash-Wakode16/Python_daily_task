# Data Cleaning

```python
import pandas as pd
```

```python
df=pd.read_csv("C:\\Users\\ywako\\Downloads\\Customer_datasets - Customer_datasets.csv")
df
```

#### 1. How many missing values are there in each column? Which column has the highest number of missing values?

```python
missing_values = df.isna().sum()
print(missing_values)
highest_missing_column = missing_values.idxmax()
highest_missing_count = missing_values.max()
print("Column with highest missing values:", highest_missing_column)
print("Number of missing values:", highest_missing_count)
```

#### 2. What percentage of rows have at least one missing value?

```python
percentage = df.isna().any(axis=1).mean() * 100
print("Percentage of rows with at least one missing value:", percentage)
```

#### 3. Drop all rows that have more than 3 missing values. How many rows remain?

```python
df_cleaned = df.dropna(thresh=df.shape[1] - 3)
print("Rows remaining:", len(df_cleaned))
```

#### 4. How many exact duplicate rows are present in the dataset?

```python
duplicate_rows = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_rows)
```

#### 5. How many duplicate Customer_IDs are there? Keep only the first occurrence of each Customer_ID.

```python
duplicate_customer_ids = df["Customer_ID"].duplicated().sum()
print("Number of duplicate Customer_IDs:", duplicate_customer_ids)
df = df.drop_duplicates(subset="Customer_ID", keep="first")
print("Rows remaining:", len(df))
```

#### 6. Convert the Age column to numeric. How many values could not be converted (became NaN)?

```python
df = df.copy()
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
print("Values that could not be converted:", df["Age"].isna().sum())
```

#### 7. After converting Age to numeric, identify and remove outliers using the IQR method. How many outliers were removed?

```python
df = df.copy()
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
outliers = (df["Age"] < lower_limit) | (df["Age"] > upper_limit)
outlier_count = outliers.sum()
print("Number of outliers:", outlier_count)
df = df[~outliers].copy()
print("Rows remaining:", len(df))
```

#### 8. Replace remaining missing values in Age with the median age.

```python
median_age = df["Age"].median()
df["Age"] = df["Age"].fillna(median_age)
print("Median Age:", median_age)
print("Missing Age values:", df["Age"].isna().sum())
```
