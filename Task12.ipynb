#### Q9. Fix Age
#### The Age column contains:
#### • Valid numbers
#### • Missing values
#### • Invalid text such as unknown, N/A, etc.
#### Convert Age into a proper numeric column.
#### Invalid values should become missing values.
#### Then fill missing Age values using the median age.
#### Use: pd.to_numeric(), median(), fillna()

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

median_age = df["Age"].median()

df["Age"] = df["Age"].fillna(median_age)
df["Age"].isnull().sum()
```

#### Q10. Fix Annual Income
#### Clean Annual_Income.
#### Requirements:
#### 1. Convert invalid values to missing.
#### 2. Convert the column to a numeric datatype.
#### 3. Calculate the median.
#### 4. Fill missing values using the median.
#### Use: pd.to_numeric(), median(), fillna(), astype()

```python
import pandas as pd
```

```python
df["Annual_Income"] = pd.to_numeric(df["Annual_Income"],errors="coerce")
df["Annual_Income"] = df["Annual_Income"].astype(float)

median_income = df["Annual_Income"].median()
print("Median Annual Income:", median_income)
df["Annual_Income"] = df["Annual_Income"].fillna(median_income)

df["Annual_Income"].isnull().sum()
df["Annual_Income"].dtype
```

#### Part 3 — Numerical Cleaning
#### Q11. Clean Account Balance
#### The Account_Balance column contains invalid strings and missing values.
#### Clean the column and replace missing values with the mean Account Balance.
#### Use:• pd.to_numeric()
#### • mean()
#### • fillna()

```python
df["Account_Balance"] = pd.to_numeric(df["Account_Balance"],errors="coerce")

mean_balance = df["Account_Balance"].mean()
print("Mean Account Balance:", mean_balance)

df["Account_Balance"] = df["Account_Balance"].fillna(mean_balance)

print("Missing values:", df["Account_Balance"].isnull().sum())
print("Data type:", df["Account_Balance"].dtype)
```

#### Q12. Clean Credit Score
#### Clean Credit_Score.
#### Invalid values should become missing.
#### Then replace missing values with the median Credit Score.
#### Finally convert the column to integer.
#### Use:
#### • pd.to_numeric()
#### • median()
#### • fillna()
#### • astype()

```python
df["Credit_Score"] = pd.to_numeric(df["Credit_Score"],errors="coerce")

median_score = df["Credit_Score"].median()
print("Median Credit Score:", median_score)

df["Credit_Score"] = df["Credit_Score"].fillna(median_score)
df["Credit_Score"] = df["Credit_Score"].astype(int)

print(df["Credit_Score"].isnull().sum())
print(df["Credit_Score"].dtype)
```

#### Q13. Clean Monthly Spend
#### Clean Monthly_Spend.
#### Then:
#### • Convert it to numeric.
#### • Replace invalid values.
#### • Replace missing values using the median.
#### • Make sure the final column is numeric.

```python
df["Monthly_Spend"] = pd.to_numeric(df["Monthly_Spend"],errors="coerce")

median_spend = df["Monthly_Spend"].median()
print("Median Monthly Spend:", median_spend)

df["Monthly_Spend"] = df["Monthly_Spend"].fillna(median_spend)

print("Missing values:", df["Monthly_Spend"].isnull().sum())
print("Data type:", df["Monthly_Spend"].dtype)
```

#### Q14. Clean Transactions
#### The Transactions column contains invalid strings.
#### Convert it to numeric and replace invalid values with the median number of transactions.
#### Then convert the final column to integer.

```python
df["Transactions"] = pd.to_numeric(df["Transactions"],errors="coerce")

median_transactions = df["Transactions"].median()
print("Median Transactions:", median_transactions)

df["Transactions"] = df["Transactions"].fillna(median_transactions)
df["Transactions"] = df["Transactions"].astype(int)

print("Missing values:", df["Transactions"].isnull().sum())
print("Data type:", df["Transactions"].dtype)
```

#### Q15. Clean Satisfaction Score
#### Clean Satisfaction_Score.
#### Replace missing/invalid values using the mean Satisfaction Score.
#### Then convert the column to float.

```python
df["Satisfaction_Score"] = pd.to_numeric(df["Satisfaction_Score"],errors="coerce")

mean_score = df["Satisfaction_Score"].mean()
print("Mean Satisfaction Score:", mean_score)

df["Satisfaction_Score"] = df["Satisfaction_Score"].fillna(mean_score)

df["Satisfaction_Score"] = df["Satisfaction_Score"].astype(float)

print("Missing values:", df["Satisfaction_Score"].isnull().sum())
print("Data type:", df["Satisfaction_Score"].dtype)
```

#### Q18. Handle missing dates
#### After conversion, find how many missing dates remain.
#### Then replace missing dates appropriately using a reasonable date-based strategy.

```python
print("Missing dates before filling:",df["Join_Date"].isnull().sum())

median_date = df["Join_Date"].dropna().median()
print("Median Join Date:", median_date)

df["Join_Date"] = df["Join_Date"].fillna(median_date)

print("Missing dates after filling:",df["Join_Date"].isnull().sum())
```

#### Part 5 — Categorical Cleaning
#### Q19. Clean City
#### The City column contains values such as:
#### Mumbai
#### mumbai
#### Mumbai
#### mumbai
#### Standardize the values so that logically identical cities are represented consistently.
#### Then check the unique frequency distribution.
#### Use: replace(), value_counts()

```python
df["City"] = df["City"].replace({
    "mumbai": "Mumbai",
    "MUMBAI": "Mumbai",
    "delhi": "Delhi",
    "DELHI": "Delhi",
    "pune": "Pune",
    "PUNE": "Pune"
})

print(df["City"].value_counts())
```

#### Q20. Analyze Segment
#### Check the frequency of:
#### Retail
#### SME
#### Corporate
#### Premium
#### Also determine whether missing values exist.
#### Then replace missing Segment values with the most appropriate category based on your analysis.
#### Use: value_counts(), fillna()

```python
print("Segment frequency:")
print(df["Segment"].value_counts(dropna=False))

print("\nMissing values:",df["Segment"].isnull().sum())

most_common_segment = df["Segment"].value_counts().idxmax()
print("Most common Segment:", most_common_segment)

df["Segment"] = df["Segment"].fillna(most_common_segment)
print("\nMissing values after filling:",
      df["Segment"].isnull().sum())
```
