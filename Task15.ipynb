#### 3. Display all column names and identify which column names contain leading/trailing spaces or inconsistent capitalization.  


```python
for col in df.columns:
    problems = []

    if col != col.strip():
        problems.append("leading/trailing spaces")

    if col != col.lower():
        problems.append("capitalization")

    if problems:
        print(col, "->", ", ".join(problems))
```

#### 10.Find all unique values in the customer_segment column and identify  inconsistent categories such as Consumer, consumer, consumer , etc.  


```python
for value in df["customer_segment"].dropna().unique():
    if value != value.strip() or value != value.lower():
        print(repr(value))
```

#### 11.Clean the customer_segment column by standardizing capitalization and  whitespace.

```python
df["customer_segment"] = (
    df["customer_segment"]
    .str.strip()
    .str.title()
)
df["customer_segment"] 
```

#### 12.Find all unique values in the gender column and convert them into consistent  categories such as Male and Female.

```python
df["gender"].str.strip().str.lower().str.title()
```

#### 13.Clean the is_active column so values such as Yes, yes, Y, 1 become 1, and  No, no, N, 0 become 0.  

```python
df["is_active"].astype(str).str.strip().str.lower().replace({
    "yes":1,
    "y":1,
    "l":1,
    "no":0,
    "n":0,
    "0":0
})
```

#### 14.Convert the churn_flag column from categorical values (Yes/No) into a binary  numerical column (1/0).  


```python
df["churn_flag"].replace({"yes":1,"no":0})
```

#### 15.Identify all values in income_usd that cannot be directly interpreted as  numbers.  

```python
df[pd.to_numeric(df['income_usd'],errors='coerce').isna()]
```

#### 16.Clean income_usd by removing currency symbols, commas, and invalid text,  then convert it to a numeric column using pd.to_numeric().  


```python
df['income_usd'].astype(str).str.replace("$",'',regex=False).str.replace(',','',regex=False)
```

#### 17.Clean the Age column so values such as "35 years" become numeric values.

```python
df["Age"] = df["Age"].astype(str).str.extract(r"(\d+)").astype(float)
df["Age"]
```

#### 18. Identify unrealistic age values using NumPy/Pandas and replace them with NaN.

```python
df[(df["Age"] < 18) | (df["Age"] > 100)]
```

#### 19.Clean the credit_score column by removing text such as "pts" and converting the column to numeric.

```python
df["credit Score"] = pd.to_numeric(df["credit Score"].astype(str).str.replace("pts", "", regex=False).str.strip(),errors="coerce")
df["credit Score"]
```
