#### 18.Extract only the time portion from Arrival_Time.

```python
df["Arrival_Time"]=pd.to_datetime(df["Arrival_Time"]).dt.time
df["Arrival_Time"]
```

#### 19.Extract the arrival hour and minute into Arrival_Hour and Arrival_Minute.

```python
df["Arrival_Hour"]=df["Arrival_Time"].apply(lambda x: x.hour)
df["Arrival_Minutes"]=df["Arrival_Time"].apply(lambda x: x.minute)
df["Arrival_Hour"]
df["Arrival_Minutes"]
```

#### 20.Drop the original Arrival_Time column after extracting the required features.

```python
df=df.drop("Arrival_Time", axis=1)
df.columns
```

#### 21.Extract the hour component from Duration and create Duration_Hour. Handle invalid or missing extracted values appropriately.

```python
df["Duration_Hour"]=(df["Duration"].str.extract(r"(\d+)\s*h", expand=False).astype("float"))
df["Duration_Hour"]
```

#### 22.Extract the minute component from Duration and create Duration_Minute. Handle invalid or missing extracted values appropriately.

```python
df["Duration_Minute"]=(df["Duration"].str.extract(r"(\d+)\s*m", expand=False).astype("float"))
df["Duration_Minute"]
```

#### 23.Drop the original Duration column after creating the numerical duration features.

```python
df=df.drop("Duration", axis=1)
df.columns
```

#### 24.Replace non-stop in Total_Stops with 0 stops.

```python
df["Total_Stops"]=df["Total_Stops"].replace("non-stop", 0)
df["Total_Stops"]
```

#### 25.Extract only the numerical value from Total_Stops and convert the column into an integer data type. Handle missing or invalid values appropriately.

```python
df["Total_Stops"] = df["Total_Stops"].str.extract(r"(\d+)")[0]

df["Total_Stops"] = pd.to_numeric(df["Total_Stops"], errors="coerce")

df["Total_Stops"] = df["Total_Stops"].fillna(0)

df["Total_Stops"] = df["Total_Stops"].astype(int)

print(df["Total_Stops"].head())
print(df["Total_Stops"].dtype)
```
