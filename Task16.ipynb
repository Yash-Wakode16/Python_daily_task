## Data Type & Invalid Values

#### 11. The Age column contains numbers as well as values like "unknown" and "twenty five". Clean this column and convert it into a proper numeric column.

```python
df['Age'] = df['Age'].astype(str).str.strip().str.lower()
df['Age'] = df['Age'].replace({'unknown': None,'twenty five': 25})
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
print(df['Age'])
```

#### 12.The Annual Income column contains values such as 650000 INR, ₹720000,1.2L, and N/A. Clean the column and convert the valid values into numeric format.

```python
df['Annual Income'] = df['Annual Income'].astype(str).str.strip().str.lower()
df['Annual Income'] = df['Annual Income'].replace({'unknown': None,'twenty five': 25})
df['Annual Income'] = pd.to_numeric(df['Annual Income'], errors='coerce')
print(df['Annual Income'])
```

#### 13.Clean the Credit Score column and identify values that cannot be converted into numbers.

```python
df['credit Score'] = df['credit Score'].astype(str).str.strip().str.lower()
df['credit Score'] = df['credit Score'].replace({'unknown': None,'twenty five': 25})
df['credit Score'] = pd.to_numeric(df['credit Score'], errors='coerce')
print(df['credit Score'])
```

#### 14.The Loan Amount column contains currency symbols, text, and numeric values. Remove the unwanted characters and convert it into numeric format.

```python
df['Loan Amount (INR)'] = (df['Loan Amount (INR)'].astype(str).str.replace(r'[₹$,]', '', regex=True).str.replace(r'[A-Za-z]', '', regex=True).str.strip())
df['Loan Amount (INR)'] = pd.to_numeric(df['Loan Amount (INR)'], errors='coerce')
df['Loan Amount (INR)'] 
```

#### 15.The Monthly Visits column contains values such as "five", "N/A", and "-". Convert valid values into numeric values and handle invalid values.

```python
df['Monthly Visits'] = df['Monthly Visits'].replace({'five': 5,'n/a': None,'-': None})
df['Monthly Visits'] = pd.to_numeric(df['Monthly Visits'],errors='coerce')
df['Monthly Visits'] 
```

#### 16. Clean the Debt to Income column where some values are written like 35% and 0.4x. Convert the valid values into a consistent numeric format.

```python
df['debt_to_income'] = df['debt_to_income'].replace({'five': 5,'n/a': None,'-': None})
df['debt_to_income'] = pd.to_numeric(df['debt_to_income'],errors='coerce')
df['debt_to_income']
```
