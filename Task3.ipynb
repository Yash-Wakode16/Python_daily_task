 ####  1. Class Name: BankAccount

#### Create a class named BankAccount.
#### Explanation
#### A bank account stores money. The balance should not be directly accessible from outside
#### the class, so create a private attribute named __balance.
#### Implement the following methods:
#### • deposit(amount)
#### o Deposit money into the account.
#### o Amount must be greater than 0.
#### • withdraw(amount)
#### o Withdraw money.
#### o Withdrawal should happen only if:
#### ▪ amount > 0
#### ▪ amount <= available balance

```python
class BankAccount:
    def __init__(self,balance):
        self.__balance=balance

    def deposit(self,amount):
        if amount>0:
            self.__balance+=amount
            print(f"{amount} deposited successfully")
        else:
            print("Invalid deposit amount")

    def withdraw(self,amount):
        if amount<=0:
            print("Invalid withdrawl amount")
        elif amount> self.__balance:
            print("Insufficient balance")
        else:
            self.__balance-=amount
            print(f"{amount} withdraw successfully")

    def show_balance(self):
            print("Current Balance :",self.__balance)

account=BankAccount(100000)
account.show_balance()

account.deposit(5000)
account.show_balance()

account.withdraw(2000)
account.show_balance()

account.withdraw(4000)

account.deposit(-100)
```

#### 10. Class Name: Account
#### Problem
#### Create class Account.
#### Explanation
#### Private attribute
#### __balance
#### Whenever balance changes,
#### store every balance value inside a class variable
#### Example
#### history = []
#### Methods
#### • deposit()
#### • withdraw()
#### Every successful update should append the latest balance to history.

```python
class Account:
    history = []

    def __init__(self, balance):
        self.__balance = 0

        if balance >= 0:
            self.__balance = balance
            Account.history.append(self.__balance)
        else:
            print("Invalid Initial Balance")

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            Account.history.append(self.__balance)
            print("Deposit Successful")
        else:
            print("Invalid Deposit Amount")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            Account.history.append(self.__balance)
            print("Withdrawal Successful")
        else:
            print("Insufficient Balance or Invalid Amount")

    def get_balance(self):
        return self.__balance


acc = Account(1000)

print("Current Balance:", acc.get_balance())

acc.deposit(500)
print("Balance:", acc.get_balance())

acc.withdraw(300)
print("Balance:", acc.get_balance())

acc.withdraw(2000)

print("Balance History:", Account.history)
```

#### 16. Class Name: Shape
#### Problem
#### Create an abstract class Shape.
#### Explanation
#### Declare abstract methods
#### • calculate_area()
#### • calculate_perimeter()
#### Create subclasses like
#### • Circle
#### • Rectangle
#### Implement both methods.

```python
from abc import ABC, abstractmethod
class Shape(ABC):

    @abstractmethod
    def calculate_area(self):
        pass

    @abstractmethod
    def calculate_perimeter(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * self.radius * self.radius

    def calculate_perimeter(self):
        return 2 * 3.14 * self.radius

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width

    def calculate_perimeter(self):
        return 2 * (self.length + self.width)

c = Circle(7)
print("Circle Area:", c.calculate_area())
print("Circle Perimeter:", c.calculate_perimeter())

r = Rectangle(8,5)
print("Rectangle Area:", r.calculate_area())
print("Rectangle Perimeter:", r.calculate_perimeter())
```

#### calculate_interest()
#### Implement in
#### • SavingsAccount

#### Each account type should use a different interest calculation.

```python
from abc import ABC, abstractmethod

class BankAccount(ABC):

    @abstractmethod
    def calculate_interest(self, balance):
        pass

class SavingsAccount(BankAccount):
    def calculate_interest(self, balance):
        interest = balance * 0.05    
        return interest

class CurrentAccount(BankAccount):
    def calculate_interest(self, balance):
        interest = balance * 0.02   
        return interest

saving = SavingsAccount()
current = CurrentAccount()

print("Savings Account Interest:", saving.calculate_interest(100000))
print("Current Account Interest:", current.calculate_interest(100000))
```

####  compute(a, b)
####  Implement in
####  • AddCalculator

```python
from abc import ABC, abstractmethod

class Calculator(ABC):
    @abstractmethod
    def compute(self, a, b):
        pass

class AddCalculator(Calculator):
    def compute(self, a, b):
        return a + b

class MultiplyCalculator(Calculator):
    def compute(self, a, b):
        return a * b

add = AddCalculator()
multiply = MultiplyCalculator()

print("Addition:", add.compute(20,10))
print("Multiplication:", multiply.compute(10,5))
```
