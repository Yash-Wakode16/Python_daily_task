# While Loop

#### Use a while loop to print numbers from 1 to 10.

```python
count = 1
while count <= 10:
    print(count)
    count += 1
```

#### Use a while loop to print all even numbers from 1 to 20.

```python
num = 1
while num <= 20:
    if num % 2 == 0:
        print(num)
    num += 1
```

#### Use a while loop to print all odd numbers from 1 to 20.

```python
num = 1
while num <= 20:
    if num % 2 != 0:
        print(num)
    num += 1
```

#### Use a while loop to print numbers from 10 down to 1.

```python
num = 10
while num >= 1:
    print(num)
    num -= 1
```

#### Take a number from the user and print its multiplication table from 1 to 10.

```python
num = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(num, "x", i, "=", num * i)
    i += 1
```

#### Take a number n from the user and calculate the sum of numbers from 1 to n.

```python
n = int(input("Enter a number: "))
i = 1
total = 0
while i <= n:
    total += i
    i += 1
print("Sum =", total)
```

#### Take an integer from the user and count how many digits it contains.

```python
num = int(input("Enter an integer: "))
count = 0
while num != 0:
    num = num // 10
    count += 1
print("Number of digits:", count)
```

#### Take an integer from the user and reverse it using a while loop.

```python
num = int(input("Enter an integer: "))
reverse = 0
while num != 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
print("Reversed number:", reverse)
```

#### Take a number and calculate the sum of its digits.

```python
num = int(input("Enter a number: "))
total = 0
while num != 0:
    digit = num % 10
    total += digit
    num = num // 10
print("Sum of digits:", total)
```

#### Take a number from the user and calculate its factorial using a while loop.

```python
num = int(input("Enter a number: "))
factorial = 1
i = 1
while i <= num:
    factorial = factorial * i
    i += 1
print("Factorial:", factorial)
```

#### Take a number and check whether it is a palindrome.

```python
num = int(input("Enter a number: "))
original = num
reverse = 0
while num != 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

#### Take an integer and find the largest digit using a while loop.

```python
num = int(input("Enter an integer: "))
largest = 0
while num != 0:
    digit = num % 10
    if digit > largest:
        largest = digit
    num = num // 10
print("Largest digit:", largest)
```

#### Take a string from the user and use a while loop to count the number of vowels.

```python
text = input("Enter a string: ")
i = 0
count = 0
while i < len(text):
    if text[i].lower() in "aeiou":
        count += 1
    i += 1
print("Number of vowels:", count)
```

#### Set a secret number, such as 7. Keep asking the user to guess the number until they enter the correct number.

```python
secret = 7
guess = int(input("Guess the number: "))
while guess != secret:
    print("Wrong guess! Try again.")
    guess = int(input("Guess the number: "))
print("Correct! You guessed the number.")
```
