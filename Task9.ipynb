####  Problem 6
####  Create a custom iterator that prints numbers from 1 to 10.

```python
class Numbers:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 10:
            value = self.num
            self.num += 1
            return value
        else:
            raise StopIteration
obj = Numbers()

for i in obj:
    print(i)
```

#### Problem 8
#### Create a custom iterator that prints numbers in reverse order from 10 to 1.

```python
class ReverseNumbers:
    def __init__(self):
        self.num = 10

    def __iter__(self):
        return self

    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num -= 1
            return value
        else:
            raise StopIteration
obj = ReverseNumbers()

for i in obj:
    print(i)
```

#### Problem 9
#### Create a custom iterator that prints the characters of the word "GENERATOR" one by one.

```python
class WordIterator:
    def __init__(self, word):
        self.word = word
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.word):
            ch = self.word[self.index]
            self.index += 1
            return ch
        else:
            raise StopIteration
obj = WordIterator("GENERATOR")

for ch in obj:
    print(ch)
```

# Generator Problems

#### Problem 11 : Create a generator that yields numbers from 1 to 10.

```python
def numbers():
    for i in range(1, 11):
        yield i

gen = numbers()

for num in gen:
    print(num)
```

#### Problem 12 : Create a generator that yields even numbers from 2 to 20.

```python
def even_numbers():
    for i in range(2, 21, 2):
        yield i
gen = even_numbers()

for num in gen:
    print(num)
```

#### Problem 13 : Create a generator that yields odd numbers from 1 to 19.

#### Problem 14 : Create a generator that yields the square of numbers from 1 to 10.

# Generator with Lists

#### Problem 16 : Given the following list:
#### numbers = [10, 15, 20, 25, 30, 35]
#### Create a generator that yields only the even numbers.

```python
numbers = [10, 15, 20, 25, 30, 35]

def even_numbers(nums):
    for num in nums:
        if num % 2 == 0:
            yield num

gen = even_numbers(numbers)

for num in gen:
    print(num)
```

#### Problem 17 : Create a generator that yields only the odd numbers from the list below.
#### numbers = [1, 4, 7, 10, 13, 16, 19]

```python
numbers = [1, 4, 7, 10, 13, 16, 19]

def odd_numbers(nums):
    for num in nums:
        if num % 2 != 0:
            yield num

gen = odd_numbers(numbers)

for value in gen:
    print(value)
```

#### Problem 18 : Given a list of names, create a generator that yields one name at a time.
#### names = ["Alice", "Bob", "Charlie", "David", "Emma"]

```python
names = ["Yash", "Ankita", "Roshan", "Rohit", "Nikhil"]

def name_generator(name_list):
    for name in name_list:
        yield name

gen = name_generator(names)

for name in gen:
    print(name)
```

#### Problem 19 : Given a list of numbers, create a generator that yields only numbers greater than 50.

#### numbers = [30, 55, 44, 98, 66, 15, 85]

```python
numbers = [30, 55, 44, 98, 66, 15, 85]

def greater_than_50(nums):
    for num in nums:
        if num > 50:
            yield num

gen = greater_than_50(numbers)

for num in gen:
    print(num)
```

#### Problem 20 : Create a generator that yields the length of each word in the following list.
#### words = ["Python", "Java", "AI", "Generator", "Iterator"]

```python
words = ["Python", "Java", "AI", "Generator", "Iterator"]

def word_lengths(word_list):
    for word in word_list:
        yield len(word)

gen = word_lengths(words)

for length in gen:
    print(length)
```

#### Problem 23 : Create a custom iterator that prints all alphabet letters from A to Z.

```python
class AlphabetIterator:
    def __init__(self):
        self.letter = 'A'

    def __iter__(self):
        return self

    def __next__(self):
        if self.letter <= 'Z':
            value = self.letter
            self.letter = chr(ord(self.letter) + 1)
            return value
        else:
            raise StopIteration

alphabets = AlphabetIterator()
for letter in alphabets:
    print(letter)
```

#### Problem 24 : Create a custom iterator that prints the Fibonacci sequence up to 10 terms.

```python
class FibonacciIterator:
    def __init__(self):
        self.a = 0
        self.b = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < 10:
            value = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return value
        else:
            raise StopIteration

fib = FibonacciIterator()
for num in fib:
    print(num)
```
