#### 1. Write a program to find the first non-repeating character in a string.

```python
text = input("Enter a string: ")
for char in text:
    if text.count(char) == 1:
        print("First non-repeating character:", char)
        break
else:
    print("No non-repeating character found")
```

#### 2. Write a program to count the frequency of each word in a sentence.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1
for word, count in frequency.items():
    print(word, ":", count)
```

#### 3. Write a program to check whether two strings are anagrams of each other.

```python
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")
if sorted(str1) == sorted(str2):
    print("Anagrams")
else:
    print("Not Anagrams")
```

#### 4. Write a program to remove duplicate characters from a string while maintaining their original order.

```python
text = input("Enter a string: ")
result = ""
for char in text:
    if char not in result:
        result += char
print("String after removing duplicates:", result)
```

#### 5. Write a program to find the longest word in a sentence.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
longest = ""
for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest word:", longest)
```

#### 6. Write a program to find the second most frequent character in a string.

```python
text = input("Enter a string: ")
frequency = {}
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1
counts = sorted(set(frequency.values()), reverse=True)
if len(counts) < 2:
    print("Second most frequent character does not exist")
else:
    second_frequency = counts[1]
    for char in text:
        if frequency[char] == second_frequency:
            print("Second most frequent character:", char)
            break
```

#### 7. Write a program to reverse each word in a sentence without changing the word order.

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
result = []
for word in words:
    result.append(word[::-1])
print("Result:", " ".join(result))
```

#### 8. Write a program to check whether a string is a palindrome, ignoring spaces and special characters.

```python
text = input("Enter a string: ")
cleaned = ""
for char in text:
    if char.isalnum():
        cleaned += char.lower()
if cleaned == cleaned[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

#### 9. Write a program to find all duplicate characters in a string.

```python
text = input("Enter a string: ")
duplicates = ""
for char in text:
    if text.count(char) > 1 and char not in duplicates:
        duplicates += char
print("Duplicate characters:", duplicates)
```

#### 10. Write a program to compress a string.


```python
text = input("Enter a string: ")
result = ""
count = 1
for i in range(len(text)):
    if i + 1 < len(text) and text[i] == text[i + 1]:
        count += 1
    else:
        result += text[i] + str(count)
        count = 1
print("Compressed string:", result)
```
