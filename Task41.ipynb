#### 1. Write a program to count vowels and consonants in a string.

```python
text=input("Enter a string: ")
vowels=0
consonants=0
for ch in text:
    if ch.isalpha():
        if ch.lower() in "aeiou":
            vowels+=1
        else:
            consonants+=1
print("Vowels:",vowels)
print("Consonants:",consonants)
```

#### 2. Write a program to count how many times each character appears in a string.

```python
text=input("Enter a string: ")
count={}
for ch in text:
    if ch in count:
        count[ch]+=1
    else:
        count[ch]=1
print(count)
```

#### 3. Given a list of numbers, find the largest and smallest element without using max() or min().

```python
numbers=[10,25,5,40,15]
largest=numbers[0]
smallest=numbers[0]
for num in numbers:
    if num>largest:
        largest=num
    if num<smallest:
        smallest=num
print("Largest:",largest)
print("Smallest:",smallest)
```

#### 4. Given a list, find the second-largest element without using sort().

```python
numbers=[10,25,5,40,15]
largest=numbers[0]
second_largest=None
for num in numbers:
    if num>largest:
        second_largest=largest
        largest=num
    elif num!=largest and (second_largest is None or num>second_largest):
        second_largest=num
print("Second Largest:",second_largest)
```

#### 5. Given two lists, find the common elements without using set().

```python
list1=[10,20,30,40,50]
list2=[30,40,50,60,70]
common=[]
for num in list1:
    if num in list2 and num not in common:
        common.append(num)
print("Common elements:",common)
```

#### 6. Write a program to remove duplicate elements from a list without using set().

```python
numbers=[10,20,10,30,20,40,30]
unique=[]
for num in numbers:
    if num not in unique:
        unique.append(num)
print("List after removing duplicates:",unique)
```

#### 7. Write a program to find the GCD of two numbers without using math.gcd().

```python
num1=int(input("Enter first number: "))
num2=int(input("Enter second number: "))
while num2!=0:
    num1,num2=num2,num1%num2
print("GCD:",num1)
```

#### 8. Write a program to check whether a number is prime.

```python
num=int(input("Enter a number: "))
if num<2:
    print("Not Prime")
else:
    prime=True
    for i in range(2,num):
        if num%i==0:
            prime=False
            break
    if prime:
        print("Prime")
    else:
        print("Not Prime")
```

#### 9. Write a program to print all prime numbers between 1 and 100.

```python
for num in range(2,101):
    prime=True
    for i in range(2,num):
        if num%i==0:
            prime=False
            break
    if prime:
        print(num)
```

#### 10. Write a program that takes a sentence and finds the most frequently occurring word.

```python
sentence=input("Enter a sentence: ")
words=sentence.lower().split()
count={}
for word in words:
    if word in count:
        count[word]+=1
    else:
        count[word]=1
most_frequent=words[0]
for word in count:
    if count[word]>count[most_frequent]:
        most_frequent=word
print("Most frequent word:",most_frequent)
print("Count:",count[most_frequent])
```
