# Method Overloading

#### 1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method
#### name.
#### Concept
#### Method Overloading (Python uses default arguments or *args to achieve overloading.)
#### What to do
#### • Create a class Calculator.
#### • Create one method add().
#### • The method should work for either 2 numbers or 3 numbers.
#### How to solve
#### 1. Create class Calculator.
#### 2. Define method add().
#### 3. Use default arguments or *args.
#### 4. Check how many numbers are provided.
#### 5. Return the sum.

```python
class Calculator:

    def add(self, *args):
        if len(args) == 2:
            return args[0] + args[1]

        elif len(args) == 3:
            return args[0] + args[1] + args[2]

        else:
            return "Please provide 2 or 3 numbers"

calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```

#### 2. Create an Area class that calculates the area of a square and rectangle using the same
#### method.
#### Concept
#### Method Overloading using default arguments.
#### What to do
#### • One parameter → Square.
#### • Two parameters → Rectangle.
#### How to solve
#### 1. Create class Area.
#### 2. Define method calculate().
#### 3. If only one value is provided, calculate the square's area.
#### 4. If two values are provided, calculate the rectangle's area.
#### 5. Return the result.

```python
class Area:

    def calculate(self, length, width=None):

        if width is None:
            return length * length

        else:
            return length * width

a = Area()
print("Area of Square:", a.calculate(5))
print("Area of Rectangle:", a.calculate(10, 5))
```

#### 3. Create a Student class with a details() method that accepts different numbers of arguments.
#### Concept
#### Method Overloading
#### What to do
#### The method should work with:
#### • Name
#### • Name + Age
#### • Name + Age + City
#### How to solve
#### 1. Create class Student.
#### 2. Create method details().
#### 3. Use default arguments.
#### 4. Print the information based on the provided values.

```python
class Student:

    def details(self, name, age=None, city=None):

        print("Name:", name)
        if age is not None:
            print("Age:", age)

        if city is not None:
            print("City:", city)

s = Student()
s.details("Yash")
print()
s.details("Rohit", 21)
print()
s.details("Nikhil", 22, "Pune")
```

#### 4. Create a MathOperations class that multiplies any number of values.
#### Concept
#### Method Overloading using *args
#### What to do
#### Multiply:
#### • 2 numbers
#### • 3 numbers
#### • 4 numbers
#### • More numbers
#### How to solve
#### 1. Create class MathOperations.
#### 2. Define method multiply(*args).
#### 3. Initialize the answer as 1.
#### 4. Loop through all numbers.
#### 5. Multiply each number.
#### 6. Return the final product.

```python
class MathOperations:

    def multiply(self, *args):
        answer = 1

        for num in args:
            answer = answer * num

        return answer

m = MathOperations()
print("2 numbers:", m.multiply(2, 3))
print("3 numbers:", m.multiply(2, 3, 4))
print("4 numbers:", m.multiply(2, 3, 4, 5))

print("More numbers:", m.multiply(1, 2, 3, 4, 5))
```

#### 5. Create a Display class that prints different messages based on the argument type.
#### Concept
#### Method Overloading + Type Checking
#### What to do
#### Accept:
#### • Integer
#### • String
#### • List
#### How to solve
#### 1. Create class Display.
#### 2. Create method show().
#### 3. Use isinstance().
#### 4. Print different messages depending on the object's type.

```python
class Display:

    def show(self, value):

        if isinstance(value, int):
            print("Integer value:", value)

        elif isinstance(value, str):
            print("String value:", value)

        elif isinstance(value, list):
            print("List value:", value)

        else:
            print("Unsupported data type")

d = Display()
d.show(10)
d.show("Hello Python")
d.show([10, 20, 30])
```

#### 6. Create a Volume class that calculates the volume of a cube and cuboid.
#### Concept
#### Method Overloading
#### What to do
#### • One parameter → Cube
#### • Three parameters → Cuboid
#### How to solve
#### 1. Create class Volume.
#### 2. Create method calculate().
#### 3. If one value is given, calculate the cube's volume.
#### 4. Otherwise, calculate the cuboid's volume.

```python
class Volume:

    def calculate(self, length, width=None, height=None):

        if width is None and height is None:
            return length ** 3

        else:
            return length * width * height

v = Volume()
print("Cube Volume:", v.calculate(5))
print("Cuboid Volume:", v.calculate(10, 5, 4))
```

#### 7. Create a PrintData class that prints one or many values.
#### Concept
#### Method Overloading using *args
#### What to do
#### Print:
#### • One value
#### • Multiple values
#### How to solve
#### 1. Create class PrintData.
#### 2. Create method print_data(*args).
#### 3. Loop through the arguments.
#### 4. Print each value.

```python
class PrintData:

    def print_data(self, *args):

        for value in args:
            print(value)

p = PrintData()
p.print_data("Hello")
print()
p.print_data(10, 20, 30, 40)
```

# Method Overriding

#### 8. Create an Animal class with sound(). Override it in Dog and Cat.
#### Concept
#### Method Overriding
#### What to do
#### • Animal → Generic sound
#### • Dog → Bark
#### • Cat → Meow
#### How to solve
#### 1. Create parent class Animal.
#### 2. Define method sound().
#### 3. Inherit Dog and Cat.
#### 4. Override sound().
#### 5. Create objects and call the method.

```python
class Animal:

    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):

    def sound(self):
        print("Dog says: Bark")

class Cat(Animal):

    def sound(self):
        print("Cat says: Meow")

a = Animal()
d = Dog()
c = Cat()

a.sound()
d.sound()
c.sound()
```

#### 9. Create Vehicle, Car, and Bike classes.
#### Concept
#### Method Overriding
#### What to do
#### Override the start() method.
#### How to solve
#### 1. Create Vehicle.
#### 2. Add method start().
#### 3. Create Car.
#### 4. Create Bike.
#### 5. Override start() in both child classes.

```python
class Vehicle:

    def start(self):
        print("Vehicle is starting")

class Car(Vehicle):

    def start(self):
        print("Car starts with a key")

class Bike(Vehicle):

    def start(self):
        print("Bike starts with a self-start button")

v = Vehicle()
c = Car()
b = Bike()

v.start()
c.start()
b.start()
```

#### 10. Create an Employee class and override calculate_salary() in Manager.
#### Concept
#### Method Overriding
#### What to do
#### • Employee → Normal salary
#### • Manager → Salary + Bonus
#### How to solve
#### 1. Create parent class.
#### 2. Define calculate_salary().
#### 3. Inherit Manager.
#### 4. Override the method.
#### 5. Print the manager's salary.

```python
class Employee:

    def calculate_salary(self, salary):
        return salary

class Manager(Employee):

    def calculate_salary(self, salary, bonus):
        return salary + bonus

e = Employee()
m = Manager()
print("Employee Salary:", e.calculate_salary(30000))
print("Manager Salary:", m.calculate_salary(50000, 10000))
```

#### 11. Create Shape, Circle, and Rectangle classes.
#### Concept
#### Method Overriding
#### What to do
#### Override the area() method.
#### How to solve
#### 1. Create Shape.
#### 2. Add area().
#### 3. Inherit Circle.
#### 4. Inherit Rectangle.
#### 5. Override area() in each child class.

```python
class Shape:
    def area(self):
        print("Area of shape")

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius * self.radius

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

c = Circle(5)
r = Rectangle(10, 4)

print("Circle Area:", c.area())
print("Rectangle Area:", r.area())
```

#### 12. Create BankAccount, SavingsAccount, and CurrentAccount.
#### Concept
#### Method Overriding
#### What to do
#### Override the interest() method.
#### How to solve
#### 1. Create BankAccount.
#### 2. Define interest().
#### 3. Override it in both child classes.
#### 4. Print different interest rates.

```python
class BankAccount:
    def interest(self):
        print("Bank Account Interest: 3%")

class SavingsAccount(BankAccount):
    def interest(self):
        print("Savings Account Interest: 4%")

class CurrentAccount(BankAccount):
    def interest(self):
        print("Current Account Interest: 2%")

bank = BankAccount()
savings = SavingsAccount()
current = CurrentAccount()
bank.interest()
savings.interest()
current.interest()
```

#### 13. Create Parent and Child classes using super().
#### Concept
#### Method Overriding + super()
#### What to do
#### Override display().
#### How to solve
#### 1. Create Parent.
#### 2. Add display().
#### 3. Create Child.
#### 4. Override display().
#### 5. Call super().display().
#### 6. Print additional information.

```python
class Parent:
    def display(self):
        print("This is Parent class")

class Child(Parent):
    def display(self):
        super().display()
        print("This is Child class")

obj = Child()
obj.display()
```

#### 14. Create Bird and Penguin classes.
#### Concept
#### Method Overriding
#### What to do
#### Override fly().
#### How to solve
#### 1. Create Bird.
#### 2. Define fly().
#### 3. Create Penguin.
#### 4. Override fly().
#### 5. Display that penguins cannot fly.

```python
class Bird:
    def fly(self):
        print("Bird can fly")

class Penguin(Bird):
    def fly(self):
        print("Penguins cannot fly")

bird = Bird()
penguin = Penguin()
bird.fly()
penguin.fly()
```
