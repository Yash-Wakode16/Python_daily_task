####  6. Class Name: BankAccount → SavingsAccount
####  Problem
####  Create a parent class named BankAccount.
####  Explanation
####  Provide methods:
####  • deposit()
####  • withdraw()
####  Create subclass:
####  • SavingsAccount
####  Add:
####  • calculate_interest()

```python
class BankAccount:
    def __init__(self,balance):
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        print("Deposited :",amount)

    def withdraw(self,amount):
        if amount<=self.balance:
            self.balance-=amount
            print("Withdraw :",amount)
        else:
            print("Insufficient balance ")

class SavingAccount(BankAccount):
    def calculate_interest(self,rate):
        interest=(self.balance*rate)/100
        print("Interest :",interest)

s1=SavingAccount(100000)

s1.deposit(3000)
s1.withdraw(4000)

s1.calculate_interest(7)

print("Current Balance :",s1.balance)
```

#### 12. Class Name: Account → LoanAccount
#### Problem
#### Create a parent class named Account.
#### Explanation
#### Store:
#### • balance
#### Create subclass:
#### • LoanAccount
#### Add:
#### • calculate_emi()

```python
class Account:
    def __init__(self,balance):
        self.balance=balance

class LoanAccount(Account):
    def __init__(self,balance,loan_amount,months):
        super().__init__(balance)
        self.loan_amount=loan_amount
        self.months=months

    def calculate_emi(self):
        emi=self.loan_amount/self.months
        print("Account Balance :",self.balance)
        print("Loan Amount :", self.loan_amount)
        print("Loan Duration :",self.months,"months")
        print("Monthly EMI :",emi)

loan = LoanAccount(70000,340000,24)
loan.calculate_emi()
```

#### 19. Class Name:
#### Shape → Polygon → Rectangle
#### Problem
#### Create multi-level inheritance.
#### Explanation
#### Rectangle should calculate its area using length and width.

```python
class Shape:
    def __init__(self):
        print("Shape Class")

class Polygon(Shape):
    def __init__(self):
        super().__init__()
        print("Polygon Class")

class Rectangle(Polygon):
    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def display(self):
        print("Length:", self.length)
        print("Width:", self.width)
        print("Area:", self.area())

r = Rectangle(8, 4)
r.display()
```

####  20. Class Name:
####  BankAccount → SavingsAccount → RecurringDeposit
####  Problem
####  Create three levels of inheritance.
####  Explanation
####  Each class should add new functionality.
####  Example:
####  • deposit
####  • interest
####  • recurring deposit maturity

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Balance after Deposit:", self.balance)

class SavingsAccount(BankAccount):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate

    def interest(self):
        interest = (self.balance * self.rate) / 100
        print("Interest:", interest)

class RecurringDeposit(SavingsAccount):
    def __init__(self, balance, rate, months):
        super().__init__(balance, rate)
        self.months = months

    def maturity(self):
        maturity_amount = self.balance + (self.balance * self.rate * self.months) / (100 * 12)
        print("Recurring Deposit Maturity Amount:", maturity_amount)

rd = RecurringDeposit(30000, 7, 24)

rd.deposit(5000)
rd.interest()
rd.maturity()
```
