#### 1. Find the most frequent element in a list.

```python
numbers=[1,2,3,2,4,2,5,3,2]
count={}
for num in numbers:
    count[num]=count.get(num,0)+1
most_frequent=max(count,key=count.get)
print("Most frequent element:",most_frequent)
```

#### 2. Find the first non-repeating character in a string.

```python
text=input()
count={}
for ch in text:
    count[ch]=count.get(ch,0)+1
for ch in text:
    if count[ch]==1:
        print("First non-repeating character:",ch)
        break
else:
    print("No non-repeating character")
```

#### 3. Find all duplicate characters in a string.

```python
text=input()
count={}
for ch in text:
    count[ch]=count.get(ch,0)+1
for ch in count:
    if count[ch]>1:
        print(ch)
```

#### 4. Count the number of vowels and consonants in a string.

```python
text=input()
vowels=0
consonants=0
for ch in text.lower():
    if ch in "aeiou":
        vowels+=1
    elif ch.isalpha():
        consonants+=1
print("Vowels:",vowels)
print("Consonants:",consonants)
```

#### 5. Reverse each word in a sentence without changing the word order.

```python
text=input()
words=text.split()
for word in words:
    print(word[::-1],end=" ")
```
