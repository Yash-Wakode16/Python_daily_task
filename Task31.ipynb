#### 1 .Take a number as input and use type() to display its data type.

```python
num = input("Enter a number: ")
print(type(num))
```

#### 2.Take two numbers and use abs() to display the absolute value of their difference.

```python
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
difference = abs(num1 - num2)
print("Absolute difference:", difference)
```

#### 3.Take a floating-point number and use int() to convert it into an integer.

```python
num = float(input("Enter a floating-point number: "))
result = int(num)
print("Integer value:", result)
```

#### 4.Take an integer and use float() to convert it into a floating-point number.

```python
num = int(input("Enter an integer: "))
result = float(num)
print("Floating-point value:", result)
```

#### 5.Take a number and use str() to convert it into a string. Print its type using type().

```python
num = int(input("Enter a number: "))
result = str(num)
print("Value:", result)
print("Type:", type(result))
```

#### 6.Take a list of numbers and use sum(), len(), and arithmetic operators to calculate the average.

```python
numbers = list(map(int, input("Enter numbers separated by space: ").split()))
total = sum(numbers)
count = len(numbers)
average = total / count
print("Sum:", total)
print("Number of elements:", count)
print("Average:", average)
```

#### 7.Take a list of numbers and use any() to check whether at least one value is True or non-zero.

```python
numbers = list(map(int, input("Enter numbers separated by space: ").split()))
result = any(numbers)
print("At least one value is non-zero:", result)
```

#### 8.Take a list of values and use all() to check whether all values are True or non-zero.

```python
values = list(map(int, input("Enter values separated by space: ").split()))
result = all(values)
print("All values are non-zero:", result)
```

#### 9.Take a string and use enumerate() to print each character along with its index.

```python
text = input("Enter a string: ")
for index, char in enumerate(text):
    print(index, char)
```

#### 10.Take two lists and use zip() to combine corresponding elements and print them.

```python
list1 = input("Enter first list elements: ").split()
list2 = input("Enter second list elements: ").split()
combined = zip(list1, list2)
for item in combined:
    print(item)
```
