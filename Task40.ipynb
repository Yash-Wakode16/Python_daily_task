#### 4. Reverse a string without using [::-1].

```python
text=input("Enter a string: ")
reverse=""

for i in text:
    reverse=i+reverse

print("Reverse:",reverse)
```

#### 5. Count how many times each character appears in a string.

```python
text=input("Enter a string: ")
count={}

for i in text:
    if i in count:
        count[i]+=1
    else:
        count[i]=1

print(count)
```

#### 6. Find the GCD of two numbers without using math.gcd().

```python
num1=int(input("Enter first number: "))
num2=int(input("Enter second number: "))

while num2!=0:
    num1,num2=num2,num1%num2

print("GCD:",num1)
```

#### 7. Find the largest and smallest element in a list without using max() or min().

```python
numbers=[10,25,5,40,15]

largest=numbers[0]
smallest=numbers[0]

for i in numbers:
    if i>largest:
        largest=i
    if i<smallest:
        smallest=i

print("Largest:",largest)
print("Smallest:",smallest)
```

#### 8. Given two lists, find the elements present in both lists without using set().

```python
list1=[1,2,3,4,5]
list2=[3,4,5,6,7]

common=[]

for i in list1:
    if i in list2:
        common.append(i)

print("Common elements:",common)
```

#### 9.Take a number and calculate its factorial.

```python
num=int(input("Enter a number: "))

fact=1

for i in range(1,num+1):
    fact*=i

print("Factorial:",fact)
```

#### 10.Take a string and count the number of vowels and consonants.

```python
text=input("Enter a string: ")

vowels=0
consonants=0

for i in text:
    if i.lower() in "aeiou":
        vowels+=1
    elif i.isalpha():
        consonants+=1

print("Vowels:",vowels)
print("Consonants:",consonants)
```
