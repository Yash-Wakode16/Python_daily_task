#### 1. Write a program to find the second largest number in a list without using sort() or max().



```python
numbers = [10, 45, 23, 89, 67, 89, 34]
largest = float('-inf')
second_largest = float('-inf')
for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num
print("Second largest number:", second_largest)
```

#### 2. Take a sentence from the user and create a dictionary containing the frequency of each word.



```python
sentence = input("Enter a sentence: ")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1
print(frequency)
```

#### 3. Given two lists, find the elements that are present in both lists without using set().

```python
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 60, 70, 80]
common = []
for item in list1:
    if item in list2:
        common.append(item)
print("Common elements:", common)
```

#### 4. Write a function is_anagram() that checks whether two strings are anagrams.

```python
def is_anagram(str1, str2):
    if len(str1) != len(str2):
        return False

    frequency = {}

    for char in str1:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

    for char in str2:
        if char not in frequency:
            return False

        frequency[char] -= 1

        if frequency[char] < 0:
            return False

    return True
str1 = input("Enter first string: ")
str2 = input("Enter second string: ")
if is_anagram(str1, str2):
    print("The strings are anagrams.")
else:
    print("The strings are not anagrams.")
```

#### 5. Using list comprehension, create a list containing the squares of only the even numbers from 1 to 20.

```python
numbers = [x**2 for x in range(1, 21) if x % 2 == 0]
print(numbers)
```
