#### 1. Given a list, find the largest and smallest element without using max() or min().

```python
numbers = [10, 25, 5, 40, 15]
largest = numbers[0]
smallest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num

    if num < smallest:
        smallest = num
print("Largest:", largest)
print("Smallest:", smallest)
```

#### 2. Given a list, calculate the sum and average of all elements.

```python
numbers = [10, 20, 30, 40, 50]
total = 0
for num in numbers:
    total = total + num
average = total / len(numbers)
print("Sum:", total)
print("Average:", average)
```

#### 3. Given a list, create a new list containing only the even numbers.

```python
numbers = [10, 15, 22, 33, 40, 51, 64]
even_numbers = []
for num in numbers:
    if num % 2 == 0:
        even_numbers.append(num)
print("Even numbers:", even_numbers)
```

#### 4. Remove duplicate elements from a list without using set(). Example: [1,2,2,3,1] → [1,2,3]

```python
numbers = [1, 2, 2, 3, 1]
unique_numbers = []
for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)
print("List without duplicates:", unique_numbers)
```

#### 5. Find the second-largest number in a list.

```python
numbers = [10, 25, 5, 40, 30]
largest = numbers[0]
second_largest = numbers[0]
for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num
print("Second-largest number:", second_largest)
```

#### 6. Given two lists, find the elements that are present in both lists without using set().

```python
list1 = [1, 2, 3, 4, 5]
list2 = [3, 4, 5, 6, 7]
common = []
for num in list1:
    if num in list2 and num not in common:
        common.append(num)
print("Common elements:", common)
```

#### 7. Separate a list into even and odd lists.

```python
numbers = [10, 15, 22, 33, 40, 51, 64]
even = []
odd = []
for num in numbers:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)
print("Even numbers:", even)
print("Odd numbers:", odd)
```

#### 8. Given a list of numbers, create a list containing their squares using list comprehension.

```python
numbers = [1, 2, 3, 4, 5]
squares = [num ** 2 for num in numbers]
print("Squares:", squares)
```

#### 9. Write a function is_prime(n) that returns True if a number is prime.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

print(is_prime(7))
print(is_prime(10))
```

#### 10. Write a function that takes a number and returns its factorial.

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result
print(factorial(5))
```
