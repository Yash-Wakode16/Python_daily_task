#### 1. Write a program to find the GCD of two numbers without using math.gcd().

```python
a=int(input())
b=int(input())
while b!=0:
    a,b=b,a%b
print("GCD:",a)
```

#### 2. Write a program to find the LCM of two numbers.

```python
a=int(input())
b=int(input())
x=a
y=b
while y!=0:
    x,y=y,x%y
gcd=x
lcm=(a*b)//gcd
print("LCM:",lcm)
```

#### 3. Write a program to count the number of digits in an integer.

```python
n=int(input())
count=0
while n!=0:
    n=n//10
    count+=1
print("Number of digits:",count)
```

#### 4. Write a program to check whether a number is an Armstrong number.

```python
n=int(input())
original=n
digits=0
temp=n
while temp!=0:
    digits+=1
    temp=temp//10
temp=n
total=0
while temp!=0:
    digit=temp%10
    total+=digit**digits
    temp=temp//10
if total==original:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

#### 5. Write a program to print a countdown from N to 1 using a while loop.

```python
n=int(input())
while n>=1:
    print(n)
    n-=1
```
