####  Challenge 1 – Complete Data Cleaning
####  Your manager gives you a messy dataset.
####  Perform all of the following:
####  • Load the dataset.
#### • Display dataset information.
####  • Find missing values.
####  • Remove duplicate rows.
####  • Fill missing values:
####  o Age → median
####  o Balance → mean
####  • Rename:
####  o Target → CustomerStatus
####  o AnnualIncome → Income
####  • Display the first 15 rows.
####  Methods expected:
####  read_csv()
####  info()
####  isnull()
####  drop_duplicates()
####  median()
####  mean()
####  fillna()
#### rename()
####  head()

```python
import pandas as pd

df=pd.read_csv("C:\\Users\\ywako\\Downloads\\Task_dataset - Task_dataset.csv")
df
```

```python
print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

df = df.drop_duplicates()

df["Age"] = df["Age"].fillna(df["Age"].median())

df["Balance"] = df["Balance"].fillna(df["Balance"].mean())

df = df.rename(columns={
    "Target": "CustomerStatus",
    "AnnualIncome": "Income"
})

df.head(15)
```

#### Challenge 2 – Banking Report
#### The bank wants a numeric report.
#### Tasks
#### • Select only numeric columns.
#### • Display summary statistics.
#### • Convert Age and CreditScore to float.
#### • Display information again.

```python
numeric_df = df.select_dtypes(include=["number"])

print("Numeric Columns:")
print(numeric_df)

print("\nSummary Statistics:")
print(numeric_df.describe())

df["Age"] = df["Age"].astype(float)
df["CreditScore"] = df["CreditScore"].astype(float)

df.info()
```

#### Challenge 3 – Duplicate Investigation
#### Find duplicate customers. Then
#### • Count duplicate rows.
#### • Remove duplicates.
#### • Print information before and after removing duplicates.
#### • Show the last 20 rows.

```python
print("Information BEFORE removing duplicates:")
df.info()

duplicates = df[df.duplicated()]

print("\nDuplicate Rows:")
print(duplicates)

duplicate_count = df.duplicated().sum()

print("\nNumber of Duplicate Rows:", duplicate_count)

df = df.drop_duplicates()

df.info()
```

#### Challenge 4 – HR Department
#### HR wants better column names.
#### Rename
#### CustomerID → Employee_ID
#### Age → EmployeeAge
#### Gender → EmployeeGender
#### AnnualIncome → Salary
#### Target → PromotionStatus
#### Then
#### • Add prefix
#### HR_
#### to every column.
#### Display 10 random records.

```python
df = df.rename(columns={
    "CustomerID": "Employee_ID",
    "Age": "EmployeeAge",
    "Gender": "EmployeeGender",
    "AnnualIncome": "Salary",
    "Target": "PromotionStatus"
})

df = df.add_prefix("HR_")

df.sample(10)
```

#### Challenge 5 – Finance Cleaning
#### Finance department reports missing values.
#### Tasks
#### Fill
#### AnnualIncome → Median
#### Balance → Mean
#### Age → Median
#### Then
#### Generate descriptive statistics.

```python
df["HR_Income"] = df["HR_Income"].fillna( df["HR_Income"].median())

df["HR_Balance"] = df["HR_Balance"].fillna(df["HR_Balance"].mean())

df["HR_EmployeeAge"] = df["HR_EmployeeAge"].fillna(df["HR_EmployeeAge"].median())

print("Descriptive Statistics:")
df.describe()
```

#### Challenge 6 – Data Type Audit
#### The company says
#### Age and AnnualIncome are stored incorrectly.
#### Tasks
#### Convert
#### Age → int
#### AnnualIncome → float
#### Balance → float
#### Target → string
#### Print information.

```python
df["Age"] = df["Age"].astype(int)
df["AnnualIncome"] = df["AnnualIncome"].astype(float)
df["Balance"] = df["Balance"].astype(float)
df["Target"] = df["Target"].astype(str)

print("Dataset Information:")
df.info()
```
