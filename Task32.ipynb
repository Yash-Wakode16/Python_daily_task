#### Print numbers from 1 to 10 using a for loop.

```python
for i in range(1, 11):
    print(i)
```

#### Print numbers from 10 to 1 using a while loop.

```python
i = 10
while i >= 1:
    print(i)
    i -= 1
```

#### Take a number from the user and print its multiplication table from 1 to 10.

```python
num = int(input("Enter a number: "))
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

#### Take a number and check whether it is a palindrome using a loop.

```python
num = int(input("Enter a number: "))
original = num
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

#### Print the following pattern:
#### *
#### **
#### ***
#### ****
#### *****

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
```
