#### 1. Survival by Gender
#### Use catplot() to show the number of passengers who survived and did not survive for each
#### gender.
#### • X-axis → sex
#### • Hue → survived
#### • Chart → count

```python
import seaborn as sns
import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")
sns.catplot(
    data=titanic,
    x="sex",
    hue="survived",
    kind="count",
    height=5,
    aspect=1.2
)
plt.title("Survival by Gender")
plt.show()
```

#### 4. Survival Percentage by Gender
#### Create a categorical visualization comparing survival across sex.
#### Use catplot() and choose an appropriate categorical plot.
#### Then identify which gender has a higher survival rate.

```python
titanic = sns.load_dataset("titanic")
survival_rate = (
    titanic.groupby("sex")["survived"]
    .mean()
    .reset_index()
)
survival_rate["survival_percentage"] = survival_rate["survived"] * 100
sns.catplot(
    data=survival_rate,
    x="sex",
    y="survival_percentage",
    kind="bar",
    height=5,
    aspect=1.2
)
plt.title("Survival Percentage by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Percentage (%)")
plt.show()
```

#### 13. Survival vs Age Group
#### Create an age_group column:
#### • Child → Age < 18
#### • Adult → 18–59
#### • Senior → 60+
#### Use catplot() to compare survival across these groups.

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt

titanic = sns.load_dataset("titanic")

titanic["age_group"] = np.select(
    [
        titanic["age"] < 18,
        titanic["age"].between(18, 59),
        titanic["age"] >= 60
    ],
    [
        "Child",
        "Adult",
        "Senior"
    ],
    default="Unknown"
)

sns.catplot(
    data=titanic,
    x="age_group",
    hue="survived",
    kind="count",
    height=5,
    aspect=1.2
)

plt.title("Survival by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Passenger Count")
plt.show()
```

#### 15. Titanic Multi-Panel Analysis
#### Create a catplot() with:
#### • X → class
#### • Y → fare
#### • Hue → sex
#### • Columns → survived
#### Use an appropriate categorical kind.
#### Add:
#### • Figure size
#### • Title
#### • Color palette
#### • Appropriate labels
#### Explain what you observe.

```python
titanic = sns.load_dataset("titanic")
g = sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    col="survived",
    kind="box",
    height=5,
    aspect=1.1,
    palette="Set2"
)
g.fig.suptitle(
    "Titanic Fare Distribution by Class, Gender and Survival",
    fontsize=16,
    fontweight="bold"
)
g.fig.subplots_adjust(top=0.82)
g.set_axis_labels("Passenger Class", "Fare")
plt.show()
```
