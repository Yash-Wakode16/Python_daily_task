#### Write a lambda function to add two numbers.

```python
add = lambda a, b: a + b
print(add(10, 20))
```

#### Write a lambda function to subtract two numbers.

```python
subtract = lambda a, b: a - b
print(subtract(20, 8))
```

#### Write a lambda function to multiply two numbers.

```python
multiply = lambda a, b: a * b
print(multiply(5, 4))
```
