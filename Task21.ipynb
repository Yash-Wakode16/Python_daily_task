# Data Cleaning

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("C:\\Users\\ywako\\Downloads\\matplotlib_messy_sales_2000 - matplotlib_messy_sales_2000.csv")
df
```

#### 1. Column Names Display all column names. Identify columns having leading/trailing spaces and rename all columns into a consistent snake_case format.

```python
print(df.columns)

for col in df.columns:
    if col != col.strip():
        print("Space found:", repr(col))
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
print(df.columns)
```

#### 2. Duplicate Records Find the number of duplicate rows. Remove the duplicate records and verify that no duplicates remain.

```python
print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()
print("Duplicates after removal:", df.duplicated().sum())
```

#### 3. Missing Values Display the missing-value count and missing-value percentage for every column.

```python
missing_count = df.isna().sum()

missing_percentage = (df.isna().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": missing_percentage
})
print(missing_summary)
```

#### 4. Missing Categorical Values Handle missing values in city, product_category, and payment_method using an appropriate strategy.

```python
df = df.copy()

df["city"] = df["city"].fillna(df["city"].mode()[0])
df["product_category"] = df["product_category"].fillna(
    df["product_category"].mode()[0]
)
df["payment_method"] = df["payment_method"].fillna(
    df["payment_method"].mode()[0]
)
print(df[["city", "product_category", "payment_method"]].isna().sum())
```

#### 5. Missing Numerical Values Handle missing values in age, unit_price, discount_pct, customer_rating, and revenue. Explain why you selected mean, median, or another method.

```python
import pandas as pd

df = df.copy()
df["age"] = pd.to_numeric(df["age"], errors="coerce")
df["unit_price"] = pd.to_numeric(df["unit_price"], errors="coerce")
df["discount_amount"] = pd.to_numeric(df["discount_amount"], errors="coerce")
df["customer_rating"] = pd.to_numeric(df["customer_rating"], errors="coerce")
df["revenue"] = pd.to_numeric(df["revenue"], errors="coerce")

df["age"] = df["age"].fillna(df["age"].median())
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].median())
df["discount_amount"] = df["discount_amount"].fillna(df["discount_amount"].median())
df["customer_rating"] = df["customer_rating"].fillna(df["customer_rating"].median())
df["revenue"] = df["revenue"].fillna(df["revenue"].median())
print(df[["age", "unit_price", "payment_method", 
          "customer_rating", "revenue"]].dtypes)
print(df[["age", "unit_price", "payment_method", 
          "customer_rating", "revenue"]].isna().sum())
```

#### 6. Date Cleaning Convert order_date into a proper datetime format. Identify invalid date values and handle them appropriately.

```python
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")

invalid_dates = df["order_date"].isna().sum()
print("Number of invalid dates:", invalid_dates)
print(df[df["order_date"].isna()])
df = df.dropna(subset=["order_date"])
print("Invalid dates after cleaning:", df["order_date"].isna().sum())
```

#### 7. Age Cleaning The age column contains values such as 25 years, 30 yrs, 45Y, and N/A. Extract the numerical age and convert the column into a numeric datatype.

```python
df["age"] = pd.to_numeric(
    df["age"].astype(str).str.extract(r"(\d+)")[0],
    errors="coerce"
)
print(df["age"].head())
print(df["age"].dtype)
```

#### 8. Age Outliers Identify unrealistic ages such as 120, 150, and 200. Detect them using the IQR method and handle them appropriately.

```python
Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
outliers = df[
    (df["age"] < lower_bound) |
    (df["age"] > upper_bound)
]
print("Age Outliers:")
print(outliers[["age"]])
```

#### 9. Unit Price Cleaning The unit_price column contains values such as ₹15000, $20000, 25000 INR, and 10k. Convert all valid values into a single numeric format.

```python
def clean_price(value):
    value = str(value).strip().lower()

    if value.endswith("k"):
        value = value[:-1]
        return float(value) * 1000

    value = value.replace("₹", "")
    value = value.replace("$", "")
    value = value.replace("inr", "")
    value = value.replace(",", "")
    return pd.to_numeric(value, errors="coerce")
df["unit_price"] = df["unit_price"].apply(clean_price)
print(df["unit_price"].head())
print(df["unit_price"].dtype)
```

#### 10. Unit Price OutliersDetect extreme unit_price values using the IQR method. Compare the number of outliers before and after treatment.

```python
import numpy as np
Q1 = df["unit_price"].quantile(0.25)
Q3 = df["unit_price"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
outliers = (
    (df["unit_price"] < lower_bound) |
    (df["unit_price"] > upper_bound)
)
before = outliers.sum()
print("Outliers before treatment:", before)
df.loc[outliers, "unit_price"] = np.nan
df["unit_price"] = df["unit_price"].fillna(df["unit_price"].median())
outliers_after = (
    (df["unit_price"] < lower_bound) |
    (df["unit_price"] > upper_bound)
)
after = outliers_after.sum()
print("Outliers after treatment:", after)
```

#### 11. Units Sold Cleaning Clean values such as 5 units, 10 pcs, twenty, N/A, and 3 and convert the column into a numeric datatype.


```python
import pandas as pd
import numpy as np

word_numbers = {
    "zero": 0,
    "one": 1,
    "two": 2,
    "three": 3,
    "four": 4,
    "five": 5,
    "six": 6,
    "seven": 7,
    "eight": 8,
    "nine": 9,
    "ten": 10,
    "eleven": 11,
    "twelve": 12,
    "thirteen": 13,
    "fourteen": 14,
    "fifteen": 15,
    "sixteen": 16,
    "seventeen": 17,
    "eighteen": 18,
    "nineteen": 19,
    "twenty": 20
}
def clean_units(value):
    value = str(value).strip().lower()
    if value in word_numbers:
        return word_numbers[value]
    value = value.replace("units", "")
    value = value.replace("unit", "")
    value = value.replace("pcs", "")
    value = value.replace("pc", "")
    return pd.to_numeric(value.strip(), errors="coerce")
df["units_sold"] = df["units_sold"].apply(clean_units)
print(df["units_sold"].head())
print(df["units_sold"].dtype)
```

#### 12. Discount Cleaning Clean values such as 10%, 20 percent, 5 %, none, and N/A. Convert the final discount_pct column into numeric values between 0 and 100.

```python
def clean_discount(value):
    value = str(value).strip().lower()
    if value in ["none", "n/a", "na", "nan", ""]:
        return np.nan
    value = value.replace("%", "")
    value = value.replace("percent", "")
    return pd.to_numeric(value.strip(), errors="coerce")
df["discount_amount"] = df["discount_amount"].apply(clean_discount)
print(df["discount_amount"].head())
print(df["discount_amount"].dtype)
```

#### 13. Discount Outliers Identify invalid discount values such as 75, 90, and 120. Determine which values are unrealistic and handle them.

```python
invalid_discounts = df[
    (df["discount_amount"] < 0) |
    (df["discount_amount"] > 100)
]
print("Invalid discount values:")
print(invalid_discounts[["discount_amount"]])
print("Number of invalid values:", len(invalid_discounts))
```

#### 15. Product Category Cleaning Standardize values such as Electronics, electronics, ELECTRONICS, Home and Kitchen, home & kitchen, and Beauty .

```python
df["product_category"] = (
    df["product_category"]
    .astype(str)
    .str.strip()
    .str.lower()
)
category_mapping = {
    "electronics": "Electronics",
    "home and kitchen": "Home & Kitchen",
    "home & kitchen": "Home & Kitchen",
    "beauty": "Beauty"
}
df["product_category"] = df["product_category"].replace(category_mapping)
print(df["product_category"].unique())
```

#### 16. Gender Cleaning Standardize values such as Male, male, Male, FEMALE, female , M, and F into consistent categories.

```python
df["gender"] = (
    df["gender"]
    .astype(str)
    .str.strip()
    .str.lower()
)
gender_mapping = {
    "male": "Male",
    "m": "Male",
    "female": "Female",
    "f": "Female"
}
df["gender"] = df["gender"].replace(gender_mapping)
print(df["gender"].unique())
```

#### 17. Customer Segment Cleaning Identify invalid values such as Prem, Reg, VIP, and Unknown. Map valid variations to the correct categories and decide how to handle invalid categories.

```python
df["customer_segment"] = (
    df["customer_segment"]
    .astype(str)
    .str.strip()
    .str.lower()
)
segment_mapping = {
    "prem": "Premium",
    "premium": "Premium",
    "reg": "Regular",
    "regular": "Regular",
    "vip": "VIP"
}
df["customer_segment"] = df["customer_segment"].replace(segment_mapping)
df["customer_segment"] = df["customer_segment"].replace(
    ["unknown", "n/a", "na", ""],
    np.nan
)
print(df["customer_segment"].unique())
```

#### 18. Binary Column Cleaning
#### Clean is_active and returned, where values contain combinations of:
#### o 0/1
#### o Yes/No
#### o Y/N
#### o Active/Inactive
#### o Returned/Not Returned
#### Convert both columns into a consistent 0/1 binary format.

```python
def clean_binary(value, column):
    value = str(value).strip().lower()
    if column == "is_active":
        mapping = {
            "1": 1,
            "yes": 1,
            "y": 1,
            "active": 1,
            "0": 0,
            "no": 0,
            "n": 0,
            "inactive": 0
        }
    else: 
        mapping = {
            "1": 1,
            "yes": 1,
            "y": 1,
            "returned": 1,
            "0": 0,
            "no": 0,
            "n": 0,
            "not returned": 0
        }

    return mapping.get(value, np.nan)
df["is_active"] = df["is_active"].apply(
    lambda x: clean_binary(x, "is_active")
)
df["returned"] = df["returned"].apply(
    lambda x: clean_binary(x, "returned")
)
print(df[["is_active", "returned"]].head())
print("is_active:", df["is_active"].unique())
print("returned:", df["returned"].unique())
```
