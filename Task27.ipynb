# Lambda function

#### Write a lambda function that takes a number and returns its square. 

```python
square = lambda x: x ** 2
print(square(8))
```

#### Create a lambda function to find the cube of a number.

```python
cube = lambda x: x ** 3
print(cube(5))
```

#### Create a lambda function that accepts two numbers and returns their sum.

```python
add = lambda a, b: a + b
print(add(10, 20))
```

#### Write a lambda function that accepts two numbers and returns the greater number.

```python
greater = lambda a, b: a if a > b else b
print(greater(10, 20))
```

#### Create a lambda function that checks whether a number is even or not.

```python
even = lambda x: x % 2 == 0
print(even(10))
```

#### Create a lambda function that checks whether a number is odd or not.

```python
odd = lambda x: x % 2 != 0
print(odd(9))
```

#### Write a lambda function to convert Celsius into Fahrenheit.

```python
fahrenheit = lambda c: (c * 9/5) + 32
print(fahrenheit(25))
```

#### Create a lambda function that calculates the final price after applying a given discount percentage.

```python
final_price = lambda price, discount: price - (price * discount / 100)
print(final_price(1000, 20))
```

#### Write a lambda function that accepts a string and returns its length.

```python
length = lambda text: len(text)
print(length("Programming"))
```

#### Create a lambda function that converts a given string into uppercase.

```python
uppercase = lambda text: text.upper()
print(uppercase("python programming"))
```

#### Write a lambda function that returns the first character of a string.

```python
first_char = lambda text: text[0]
print(first_char("Python"))
```

#### data = [(1, 5), (2, 3), (4, 1), (3, 2)]
#### Use lambda with sorted() to sort the list according to the second value.

```python
data = [(1, 5), (2, 3), (4, 1), (3, 2)]
result = sorted(data, key=lambda x: x[1])
print(result)
```
