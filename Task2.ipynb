### 28. In class Product, add getter and setter for private __price with discount logic.

```python
class Product:
    def __init__(self,price):
        self.__price=price

    def get_price(self):
        return self.__price

    def set_price(self,price):
        if price>0:
            self.__price=price
            print("Price updated successfully")
        else:
            print("Invalid Price")

    def apply_discount(self,discount):
        if 0<= discount <=100:
            discounted_price=self.__price-(self.__price*discount/100)
            self.__price=discounted_price
            print(f"{discount}%discount applied")
        else:
            print("Invalid Discount!")

p1=Product(50000)
print("Original Price :",p1.get_price())

p1.apply_discount(20)
print("Price after discount :",p1.get_price())

p1.set_price(45000)
print("New price :",p1.get_price())
            
```

### 30. Define class Rectangle with private __length and __width. Add getter/setter and area method.

```python
class Rectangle:
    def __init__(self, length, width):
        self.__length = length   
        self.__width = width     

    def get_length(self):
        return self.__length

    def get_width(self):
        return self.__width

    def set_length(self, length):
        if length > 0:
            self.__length = length
            print("Length updated successfully.")
        else:
            print("Invalid Length!")

    def set_width(self, width):
        if width > 0:
            self.__width = width
            print("Width updated successfully.")
        else:
            print("Invalid Width!")

    def area(self):
        return self.__length * self.__width


rect = Rectangle(10, 5)

print("Length:", rect.get_length())
print("Width:", rect.get_width())

print("Area:", rect.area())

rect.set_length(20)
rect.set_width(8)

print("\nUpdated Length:", rect.get_length())
print("Updated Width:", rect.get_width())

print("Updated Area:", rect.area())
```
