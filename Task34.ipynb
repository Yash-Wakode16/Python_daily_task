#### 1. Write a Python program to print numbers from 1 to 10 using a for loop.

```python
for i in range(1, 11):
    print(i)
```

#### 2. Write a program to print all even numbers from 1 to 20 using a loop.

```python
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

#### 3. Write a program to find the sum of numbers from 1 to 100 using a for loop.

```python
sum = 0
for i in range(1, 101):
    sum = sum + i
print("Sum =", sum)
```

#### 4. Take a number from the user and print its multiplication table from 1 to 10.

```python
num = int(input("Enter a number: "))
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

#### 5. Write a program to print numbers from 10 to 1 using a while loop.

```python
i = 10
while i >= 1:
    print(i)
    i = i - 1
```

#### 6. Take an integer from the user and use a while loop to count how many digits it contains.

```python
num = int(input("Enter an integer: "))
num = abs(num)
count = 0
if num == 0:
    count = 1
else:
    while num > 0:
        num = num // 10
        count = count + 1
print("Number of digits =", count)
```

#### 7. Take a number from the user and find its factorial using a loop.

```python
num = int(input("Enter a number: "))
factorial = 1
for i in range(1, num + 1):
    factorial = factorial * i
print("Factorial =", factorial)
```

#### 8. Take a number from the user and use a loop to check whether it is prime or not.

```python
num = int(input("Enter a number: "))
if num <= 1:
    print("Not Prime")
else:
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print("Prime Number")
    else:
        print("Not Prime")
```

#### 9. Take a number from the user and use a while loop to check whether the number is a palindrome.

```python
num = int(input("Enter a number: "))
original = num
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10
if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

#### 10. Write a program using nested loops to print the following pattern:

#### *
#### **
#### ***
#### ****
#### *****

```python
for i in range(1, 6):
    for j in range(i):
        print("*", end="")
    print()
```
