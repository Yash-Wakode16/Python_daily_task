#### 1. Take a list of numbers and find the second largest number without using sort().

```python
numbers = [10, 25, 5, 40, 30]
largest = second_largest = float('-inf')
for num in numbers:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print("Second largest number:", second_largest)
```

#### 2. Take a string from the user and check whether it is a palindrome.

```python
text = input("Enter a string: ")
if text == text[::-1]:
    print("Palindrome")
else:
    print("Not a palindrome")
```

#### 3. Print all prime numbers between 1 and 100.

```python
for num in range(2, 101):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

#### 4. Take a string and create a dictionary containing the frequency of each character. Example: "hello" → {'h':1, 'e':1, 'l':2, 'o':1}

```python
text = input("Enter a string: ")
frequency = {}
for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1
print(frequency)
```

#### 5. Remove duplicate elements from a list without using set().

```python
numbers = [10, 20, 10, 30, 20, 40, 30]
unique = []
for num in numbers:
    if num not in unique:
        unique.append(num)
print(unique)
```

#### 6. Given two lists, find the elements that are present in both lists without using set().

```python
list1 = [10, 20, 30, 40, 50]
list2 = [30, 40, 50, 60, 70]
common = []
for num in list1:
    if num in list2:
        common.append(num)
print(common)
```

#### 7. Using list comprehension, create a list containing the squares of even numbers from 1 to 20.

```python
squares = [num ** 2 for num in range(1, 21) if num % 2 == 0] 
print(squares)
```

#### 8. Take a sentence and create a dictionary containing the frequency of each word. Example: "python is easy python is powerful"

```python
sentence = input("Enter a sentence: ")
words = sentence.split()
frequency = {}
for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1
print(frequency)
```

#### 9. Write a function is_anagram() that checks whether two strings are anagrams. Example: "listen" and "silent" → True

```python
def is_anagram(str1, str2):
    return sorted(str1) == sorted(str2)
print(is_anagram("listen", "silent"))
```

#### 10. Write a function that takes a number and returns its factorial using a loop.

```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result = result * i
    return result
num = int(input("Enter a number: "))
print("Factorial:", factorial(num))
```
