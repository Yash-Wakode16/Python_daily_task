# Hotel Dataset

#### Hotel Dataset — 30 Data Cleaning Questions
#### 1. Remove leading/trailing spaces from column names
#### Explanation: Column names like "Booking ID " and "Customer Name" have extra spaces. This
#### causes errors when you try to select columns. Just strip the spaces so names become clean
#### (Booking ID, Customer Name).

```python
import pandas as pd
```

```python
df=pd.read_csv("C:\\Users\\ywako\\Downloads\\Hotel_booking_dataset - Hotel_booking_dataset.csv")
df
```

```python
df.columns = df.columns.str.strip()
df.columns
```

#### 2. Standardize all date formats into one consistent format (YYYY-MM-DD)
#### Explanation: Dates appear as 2025-05-17 00:00:00, 2025/02/20, 15/01/2025, and even
#### invalid ones like 32/13/2025. Convert everything to a single clean format and turn bad dates
#### into missing values.

```python
df["Booking Date"] = pd.to_datetime(df["Booking Date"], errors="coerce", dayfirst=True)
df["Booking Date"] = df["Booking Date"].dt.strftime("%Y-%m-%d")
df["Booking Date"]
```

#### 3. Handle completely missing Customer Name values
#### Explanation: Some rows have blank Customer Name or values like N/A. Decide whether to
#### fill them with a placeholder (e.g., “Unknown”) or drop those rows, depending on whether
#### the name is critical for analysis.

```python
df["Customer Name"] = df["Customer Name"].replace(["N/A", "NA", "n/a", ""],pd.NA)
df["Customer Name"] = df["Customer Name"].fillna("Unknown")
df["Customer Name"]
```

#### 4. Fix inconsistent Gender values Explanation: Gender appears as Male, FEMALE, male, F, M, Other, etc. Convert everything to a standard set such as Male, Female, Other.

```python
df["Gender"] = df["Gender"].str.strip().str.lower()
df["Gender"] = df["Gender"].replace({
    "male": "Male",
    "m": "Male",
    "female": "Female",
    "f": "Female",
    "other": "Other"
})
df["Gender"].unique()
```

#### 5. Clean and standardize City names Explanation: Cities have typos and casing issues (mumbai, Mumbai , Bangalore, Hydrabad, blank). Correct spelling, remove extra spaces, and make casing consistent (e.g., Title Case).

```python
df["City"] = df["City"].str.strip().str.lower()
df["City"] = df["City"].replace({
    "mumbai": "Mumbai",
    "bangalore": "Bangalore",
    "hydrabad": "Hyderabad"
})
df["City"] = df["City"].str.title()
df["City"] = df["City"].replace("", "Unknown").fillna("Unknown")
df["City"].unique()
```

#### 6. Standardize Room Type categories Explanation: Room Type has Suite, Deluxe, delux, standard, exec, Unknown, and blanks. Map similar values to a clean list: Standard, Deluxe, Suite, Executive.

```python
df["Room Type"] = df["Room Type"].str.strip().str.lower()
df["Room Type"] = df["Room Type"].replace({
    "standard": "Standard",
    "deluxe": "Deluxe",
    "delux": "Deluxe",
    "suite": "Suite",
    "exec": "Executive",
    "executive": "Executive",
    "unknown": "Unknown",
    "": "Unknown"
})
df["Room Type"] = df["Room Type"].fillna("Unknown")
df["Room Type"].unique()
```

#### 7. Clean Booking Channel values Explanation: Values include Online, online, Walk-in, Travel Agent, TravelAgent, corp, Corporate. Unify them into a small consistent set (Online, Walk-in, Corporate, Travel Agent).

```python
df["Booking Channel"] = df["Booking Channel"].str.strip().str.lower()
df["Booking Channel"] = df["Booking Channel"].replace({
    "online": "Online",
    "walk-in": "Walk-in",
    "walk in": "Walk-in",
    "travel agent": "Travel Agent",
    "travelagent": "Travel Agent",
    "corp": "Corporate",
    "corporate": "Corporate"
})
df["Booking Channel"].unique()
```

#### 8. Standardize Customer Segment Explanation: Segments appear as Premium, regular, REG, Budget , VIP, etc. Create a clean set such as Budget, Regular, Premium, Corporate, VIP and map everything to it.

```python
df["Customer Segment"] = df["Customer Segment"].str.strip().str.lower()
df["Customer Segment"] = df["Customer Segment"].replace({
    "budget": "Budget",
    "regular": "Regular",
    "reg": "Regular",
    "premium": "Premium",
    "corporate": "Corporate",
    "corp": "Corporate",
    "vip": "VIP"
})
df["Customer Segment"] = df["Customer Segment"].fillna("Unknown")
df["Customer Segment"].unique()
```

#### 9. Fix Age column (invalid and non-numeric values) Explanation: Age has numbers, blanks, 30 yrs, 25 years, 45Y, and impossible values like 150 or 100. Extract the number, convert to integer, and set unrealistic ages (e.g., <18 or >100) to missing.

```python
df["Age"] = df["Age"].astype(str).str.extract(r"(\d+)")[0]
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df.loc[(df["Age"] < 18) | (df["Age"] > 100), "Age"] = pd.NA
df["Age"] = df["Age"].astype("Int64")
df["Age"]
```

#### 10. Clean Room Nights column Explanation: Values include numbers, N/A, 3 nights, 100, 30. Extract only the numeric part, convert to integer, and decide what to do with extreme values (e.g., 100 nights).

```python
df["Room Nights"] = (
    df["Room Nights"]
    .astype(str)
    .str.extract(r"(\d+)")[0]
)
df["Room Nights"] = pd.to_numeric(
    df["Room Nights"],
    errors="coerce"
)
df.loc[df["Room Nights"] > 30, "Room Nights"] = pd.NA
df["Room Nights"] = df["Room Nights"].astype("Int64")
df["Room Nights"]
```

#### 11. Fix Room Rate (mixed formats and extreme values) Explanation: Room Rate contains normal numbers, 15k, $8000, 12000 INR, 100000, 250000, and blanks. Convert text formats to numbers, remove currency symbols, and flag or handle unrealistic rates.

```python
import pandas as pd
import numpy as np
df["Room Rate"] = df["Room Rate"].astype(str).str.strip().str.lower()
df["Room Rate"] = df["Room Rate"].str.replace(
    r"(\d+(?:\.\d+)?)\s*k\b",
    lambda x: str(float(x.group(1)) * 1000),
    regex=True
)
df["Room Rate"] = (
    df["Room Rate"]
    .str.replace(r"[$₹,]", "", regex=True)
    .str.replace(r"\s*inr\b", "", regex=True)
    .str.strip()
)
df["Room Rate"] = pd.to_numeric(
    df["Room Rate"],
    errors="coerce"
)
df["Room Rate Flag"] = np.where(
    (df["Room Rate"] > 50000) | (df["Room Rate"] <= 0),
    "Unrealistic",
    "Valid"
)
df.loc[df["Room Rate Flag"] == "Unrealistic", "Room Rate"] = np.nan
df[["Room Rate", "Room Rate Flag"]]
```

#### 12. Clean Discount % columnExplanation: Discount has numbers, none, 10%, 5 %, N/A, and impossible values like 120. Convert everything to a pure number between 0 and 100; treat “none” as 0 and values >100 as missing or capped.

```python
df["Discount %"] = df["Discount %"].astype(str).str.strip().str.lower()
df["Discount %"] = df["Discount %"].replace({
    "none": "0",
    "n/a": pd.NA,
    "na": pd.NA,
    "": pd.NA
})
df["Discount %"] = (
    df["Discount %"]
    .str.replace("%", "", regex=False)
)
df["Discount %"] = pd.to_numeric(
    df["Discount %"],
    errors="coerce"
)
df.loc[df["Discount %"] > 100, "Discount %"] = pd.NA
df["Discount %"]
```

#### 13. Handle missing and invalid Rating values Explanation: Rating has numbers, blanks, 5/5, 4 stars, and values above 5 (e.g., 6, 9). Convert to a 1–5 scale and set invalid entries to missing.

```python
df["Rating"] = df["Rating"].astype(str).str.strip().str.lower()
df["Rating"] = df["Rating"].str.extract(r"(\d+(?:\.\d+)?)")[0]
df["Rating"] = pd.to_numeric(
    df["Rating"],
    errors="coerce"
)
df.loc[
    (df["Rating"] < 1) | (df["Rating"] > 5),
    "Rating"
] = pd.NA
df["Rating"] = df["Rating"].astype("Float64")
df["Rating"]
```

#### 14. Clean Guests column (extreme values) Explanation: Most values are reasonable (1–6), but some are 15, 25, or 40. Decide a sensible maximum (e.g., 10) and set higher values to missing or investigate them.

```python
df["Guests"] = pd.to_numeric(
    df["Guests"],
    errors="coerce"
)
df.loc[
    (df["Guests"] < 1) | (df["Guests"] > 10),
    "Guests"
] = pd.NA
df["Guests"] = df["Guests"].astype("Int64")
df["Guests"]
```

#### 15. Standardize Is Member column Explanation: Values are mixed: 1, 0, Y, Yes, Member, blanks. Convert everything to a clear Yes/No or 1/0 format.

```python
df["Is Member"] = df["Is Member"].astype(str).str.strip().str.lower()
df["Is Member"] = df["Is Member"].replace({
    "1": "Yes",
    "y": "Yes",
    "yes": "Yes",
    "member": "Yes",
    "0": "No",
    "n": "No",
    "no": "No"
})
df["Is Member"] = df["Is Member"].replace({
    "nan": pd.NA,
    "": pd.NA
})
df["Is Member"].unique()
```

#### 16. Standardize Cancelled column Explanation: Cancelled appears as 0, 1, Yes, Y, N. Convert to a consistent 0/1 or Yes/No format.

```python
df["Cancelled"] = df["Cancelled"].astype(str).str.strip().str.lower()
df["Cancelled"] = df["Cancelled"].replace({
    "yes": 1,
    "y": 1,
    "1": 1,
    "no": 0,
    "n": 0,
    "0": 0
})
df["Cancelled"] = pd.to_numeric(
    df["Cancelled"],
    errors="coerce"
)
df["Cancelled"].unique()
```

#### 17. Clean Payment Method values Explanation: Payment methods have inconsistent casing and spacing (Credit Card, credit card, upi, Net Banking, cash , blanks). Standardize to a clean list (Credit Card, Debit Card, UPI, Net Banking, Cash).

```python
df["Payment Method"] = (
    df["Payment Method"]
    .astype(str)
    .str.strip()
    .str.lower()
)
df["Payment Method"] = df["Payment Method"].replace({
    "credit card": "Credit Card",
    "creditcard": "Credit Card",
    "debit card": "Debit Card",
    "debitcard": "Debit Card",
    "upi": "UPI",
    "net banking": "Net Banking",
    "netbanking": "Net Banking",
    "cash": "Cash"
})
df["Payment Method"] = df["Payment Method"].replace({
    "": pd.NA,
    "nan": pd.NA
})
df["Payment Method"].unique()
```

#### 18. Remove or flag duplicate Booking IDs Explanation: Some Booking IDs (e.g., BKG00042, BKG01640) appear more than once. Decide whether to keep the first occurrence, the most complete row, or mark them for review.

```python
df["Duplicate Booking ID"] = df["Booking ID"].duplicated(
    keep=False
)
duplicates = df[df["Duplicate Booking ID"]].sort_values("Booking ID")
duplicates
```

#### 19. Handle missing Room Type values Explanation: Some rows have blank Room Type. You can either drop those rows, fill with “Unknown”, or try to infer from other columns if possible.

```python
df["Room Type"] = df["Room Type"].replace(
    r"^\s*$",
    pd.NA,
    regex=True
)
df["Room Type"] = df["Room Type"].fillna("Unknown")
print(df["Room Type"].isna().sum())
df["Room Type"].unique()
```

#### 20. Fix inconsistent Final Amount calculations Explanation: Final Amount should roughly equal Gross Amount − Discount Amount, but some rows have missing Final Amount or clear calculation errors. Recalculate or flag inconsistencies.

```python
amount_columns = ["Gross Amount", "Discount Amount", "Final Amount"]
for col in amount_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
df["Expected Final Amount"] = (
    df["Gross Amount"] - df["Discount Amount"]
)
df["Amount Difference"] = (
    df["Final Amount"] - df["Expected Final Amount"]
).abs()
df["Final Amount Status"] = np.where(
    df["Final Amount"].isna(),
    "Missing",
    np.where(
        df["Amount Difference"] > 1,
        "Inconsistent",
        "Valid"
    )
)
df.loc[
    df["Final Amount Status"].isin(["Missing", "Inconsistent"]),
    "Final Amount"
] = df["Expected Final Amount"]
(
    df[
        [
            "Gross Amount",
            "Discount Amount",
            "Final Amount",
            "Expected Final Amount",
            "Final Amount Status"
        ]
    ]
)
```
